### Метод наименьших квадратов


Если система переопределена ($m>n+1$), точного решения обычно нет. Поэтому используется метод наименьших квадратов (МНК), который дает приближённое решение:

$\hat{w} = (A^{\top} A)^{-1} A^{\top} y$

Таким образом, коэффициенты линейной регрессии находятся через обращение матрицы Грама $A^{\top}A$.

### Корреляционная матрица


Она в точности совпадает с матрицей Грама, вычисленной для стандартизированных векторов $\vec{x}_{1_{st}}$ и $\vec{x}_{2_{st}}$:

$
C = G(\vec{x}_{1_{st}}, \vec{x}_{2_{st}})
$

Но «магия» ещё не закончилась. Давайте подумаем: какова геометрическая интерпретация корреляции?

Присмотритесь к формуле, вспомните свойства скалярного произведения, а затем загляните в ответ:т:

$c_{i j}=\frac{\left(\vec{x}_{i_{\text {cent }}} \vec{x}_{j_{\text {cent }}}\right)}{\left\|\vec{x}_{i_{\text {cent }}}\right\| \cdot\left\|\vec{x}_{j_{\text {cent }}}\right\|}$

Это косинус угла между центрированными векторами  и . По свойству скалярного произведения:

Основная формула коэффициента корреляции Пирсона

Чаще всего используются две математически эквивалентные формулы:

1. Ковариационная формула:

$
r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2 \sum_{i=1}^{n}(y_i - \bar{y})^2}} 
$

3. Вычислительная формула (более удобная для калькулятора):

$
r = \frac{n\sum x_i y_i - \sum x_i \sum y_i}{\sqrt{[n\sum x_i^2 - (\sum x_i)^2][n\sum y_i^2 - (\sum y_i)^2]}} 
$

Хотя эти формулы математически тождественны, они вычислительно неравноценны.
