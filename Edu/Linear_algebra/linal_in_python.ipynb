```python
Count_DF = pd.DataFrame({
    'Женские стрижки': [10, 2, 12, 4, 6, 10, 22, 7], 
    'Мужские стрижки': [5, 21, 12, 8, 25, 3, 1, 0], 
    'Окрашивания':[12, 3, 0, 18, 27, 2, 4, 31],
    'Укладка':[15, 25, 30, 14, 25, 17, 25, 31],
    'Уход':[10, 6, 4, 5, 18, 12, 20, 28]
    }, 
    index=['Аня', 'Борис', 'Вика', 'Галя', 'Дима', 'Егор', 'Женя','Юра']
)

Price_DF = pd.DataFrame({
    'Женские стрижки': [2, 1.8, 2, 1.8, 2.5, 5, 1.1, 4.5], 
    'Мужские стрижки': [1.5, 2.5, 2, 1.2, 3.5, 5, 1, 4], 
    'Окрашивания':[1, 1, 0, 2.8, 2, 3, 1.5, 2.5],
    'Укладка':[0.8, 1, 0.5, 0.8, 1, 2, 0.5, 1],
    'Уход':[1, 1, 2, 2, 1.5, 2.5, 1.7, 2] 
    }, 
    index=['Аня', 'Борис', 'Вика', 'Галя', 'Дима', 'Егор', 'Женя','Юра']
)
```

```python
Count_DF, Price_DF
```

```python
# вектор комиссий
com = np.array([0.2, 0.2, 0.3, 0.1, 0.1])
com
```

```python
m_income = m_count * m_price
m_income
```

```python
m_com = m_income * com
m_com
```

```python
v_com = m_income @ com
v_com
```

```python
# сумма по столбцам
np.sum(m_com, axis=0)
```

```python
# сумма по строкам
np.sum(m_com, axis=1)
```

```python
m_profit = m_income - m_com
m_profit
```

```python
import numpy as np

# Создаем матрицу
A = np.array([[1, 4, 13], [3, -4, 7], [5, 9, 12]])

# Вычисляем собственные числа
eigenvalues = np.linalg.eigvals(A)

print("Собственные числа матрицы A:")
print(eigenvalues)
a, b, c = eigenvalues
print(round(a), round(b), round(c))
```

```python
A = np.array([[1, 9, 4], [9, 4, 7],[4, 7, 12]])
A
```

```python
A.T
```

```python
A.T @ A
```

```python
# Вычисляем собственные числа
eigenvalues = np.linalg.eigvals(A.T @ A)

print("Собственные числа матрицы A:")
print(eigenvalues)
a, b, c = eigenvalues
print(round(a), round(b), round(c))
```
