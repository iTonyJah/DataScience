```python
def prime_num(num):
    for i_num in range (2, int(num ** (1/2))+1):
        if num % i_num == 0:
               print(f'Число {num} не простое')
               break
    else:
        print(f'число {num} простое')

user_num = 999_999_999_989
print(user_num, type(user_num))
prime_num(user_num)
```

```python
def is_prime_optimized(n):
    """
    Оптимизированная проверка на простоту
    Использует математические свойства простых чисел
    """
    if n < 2:
        return False
    if n == 2 or n == 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    
    # Все простые числа > 3 имеют вид 6k ± 1
    # Проверяем только такие делители
    i = 5
    w = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += w
        w = 6 - w  # чередуем +2 и +4: 5,7,11,13,17,19...
    
    return True

is_prime_optimized(999_999_999_989)
```

```python
def equalize_lengths(lst):

    lengths = {k: len(v) for k, v in enumerate(lst)}
    max_len = max(lengths.values())
    lengths = dict(sorted(lengths.items(), key=lambda item: item[1], reverse=True))

    equal_length = []
    for key, value in lengths.items():
        equal_length.append(lst[key] + '_' * (max_len - value))

    return equal_length
```

```python
print(equalize_lengths(['крот', 'белка', 'выхухоль']))
# ['выхухоль', 'белка___', 'крот____']

print(equalize_lengths(['a', 'aa', 'aaa', 'aaaa', 'aaaaa']))
# ['aaaaa', 'aaaa_', 'aaa__', 'aa___', 'a____']

print(equalize_lengths(['qweasdqweas', 'q', 'rteww', 'ewqqqqq']))
# ['qweasdqweas', 'ewqqqqq____', 'rteww______', 'q__________']
```

```python
def find_min_number(*args, **kwargs):
    
    number_list = list(kwargs.values())
    number_list.extend(args)
    
    min_num = None
    for num in number_list:
        if min_num is None:
            min_num = num
        elif num < min_num:
            min_num = num
    
    return min_num
```

```python
print(find_min_number(130, 122, 19))
# 19
print(find_min_number(10.9, 12.2, 18.4))
# 10.9
print(find_min_number(-0.9, -12.2, -18.4, 0, 1, 2, 3, 50, 1000000000))

print(find_min_number(5, 15, -1,a = 150, b = 360, c = 10, ))
print(find_min_number(1, 2, a = 15, c = -6, b = 50))
print(find_min_number(1))
```

```python
def sum_min_numbers(*args, **kwargs):
    
    number_list = list(kwargs.values())
    number_list.extend(args)

    sum_min = 0
    for _ in range(2):
        min_n = find_min_number(*number_list)
        number_list.pop(number_list.index(min_n))
        sum_min += min_n

    return sum_min
    
```

```python
print(sum_min_numbers(1, 2, 3))
# 3
print(sum_min_numbers(1, 2, -10))
# -9
print(sum_min_numbers(a = -1, b = 0, c = 10))

```
