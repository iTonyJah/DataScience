<b>1 задание</b>

Назовем натуральное число честно четным, если все цифры в его десятичной записи четные. Например, числа 4826 и 8802 - честно четные, а 79 и 301 - нет.

Вам необходимо проверить, является ли введенное число n честно четным.

Входные данные
Натуральное число n ≤ 10^9.

Выходные данные
Выведите число 1, если n является честно четным, и 0 - иначе.

```python
n = int(input())
is_honest_even = True

for digit in str(n):
    if int(digit) % 2 != 0:
        is_honest_even = False
        break

print(1 if is_honest_even else 0)
```

<b>3 задание</b>

Даны две рациональные дроби: a/b и c/d. Сложите их и результат представьте в виде несократимой дроби m/n.

Входные данные
Программа получает на вход 4 натуральных числа a, b, c, d, не превосходящих 100.

Выходные данные
Программа должна вывести 2 натуральных числа m и n такие, что m/n = a/b + c/d и дробь m/n несократима.

```python
import math

a, b, c, d = map(int, input().split())
m = a * d + c * b
n = b * d

# greatest common divisor (GCD), наибольший общий делитель (НОД)
gcd = math.gcd(m, n)

print(m // gcd, n // gcd)
```

<b>4 задание</b>

Вам дана последовательность A, изначально состоящая из одного элемента 0 (A=(0)) и строка S длины N, состоящая только из символов <<L>> и <<R>>.

Затем вы проделаете следующую операцию для i=1,2,3,…,N:

Если Si =L, то вы вставите число i слева от числа i−1 в последовательность A.

Если Si =R, то вы вставите число i справа от числа i−1 в последовательность A.

Найдите последовательность, которая получится после всех операций.

Входные данные
В первой строке вводится одно целое число N (1≤N≤5*10^5) — длина строки S.

Во второй строке вводится строка S (∣S∣=N). Гарантируется, что S состоит только из символов <<l>> и <<r>>.

Выходные данные
Выведите N+1 чисел через пробел — последовательность A, которая получится в результате применения N операций.

```python
n = int(input())
s = input().upper()

a = [0]
#list.insert(index, element)

for i, letter in enumerate(s):

    if letter == 'L':
        shift = 0
    else:
        shift = 1
    
    a.insert(a.index(i) + shift, i+1)

#распаковываем список
print(*a)
```

<b>4 задание</b>

Мальчик подошел к лестнице. C каждой ступенькой на ней у него ассоциирован какой-то момент из жизни, который при воспоминании изменит его текущее настроение на ai. Мальчик умеет перешагивать на следующую ступеньку, либо перепрыгивать через ступеньку. Еще мальчик умеет абстрагироваться от воспоминаний и проходить вперед на любое количество ступенек, не изменяя свое настроение (в том числе и оказаться на последней ступеньке), но так как он еще недостаточно взрослый, он может сделать это не более k раз. Какое максимальное настроение он может получить, оказавшись на последней ступеньке?

Входные данные
В первой строке вводятся два целых числа n (1≤n≤1000) и k (0≤k≤100). Во второй строке вводятся n целых чисел ai (∣ai∣≤100)  — изменение настроения, если наступить на i-ю ступеньку (снизу вверх).

Выходные данные
Выведите единственное число  — максимальное настроение, которое он может получить, оказавшись на последней ступеньке.

Примечание
Изначально мальчик стоит перед первой ступенькой (не на ней).

```python
n, k = map(int, input().split())

a = list(map(int, input().split()))
print(n, k, a)


# Инициализация DP: dp[i][k_used] = максимальное настроение на i-й ступеньке
dp = [[-10**18 for _ in range(k + 1)] for __ in range(n + 1)]
dp[0][0] = 0  # Начальное состояние перед лестницей

for i in range(n):
    for k_used in range(k + 1):
        if dp[i][k_used] == -10**18:
            continue
        
        # Вариант 1: Перейти на следующую ступеньку (+1)
        if i + 1 <= n:
            new_mood = dp[i][k_used] + (a[i] if i + 1 == n else a[i])
            if new_mood > dp[i + 1][k_used]:
                dp[i + 1][k_used] = new_mood
        
        # Вариант 2: Перепрыгнуть через одну ступеньку (+2)
        if i + 2 <= n:
            new_mood = dp[i][k_used] + (a[i] if i + 2 == n else a[i])
            if new_mood > dp[i + 2][k_used]:
                dp[i + 2][k_used] = new_mood
        
        # Вариант 3: Пропустить ступеньки (если остались пропуски)
        if k_used < k:
            # Пропускаем до предпоследней, но последнюю надо посетить
            if i < n - 1:
                new_mood = dp[i][k_used] + a[-1]  # Обязательно наступаем на последнюю
                if new_mood > dp[n][k_used + 1]:
                    dp[n][k_used + 1] = new_mood

# Ответ — максимум среди dp[n][0..k]
max_mood = max(dp[n][j] for j in range(k + 1))
print(max_mood)
```

```python
n, k = map(int, input().split())
a = list(map(int, input().split()))


s = 0
skip = False
for i, el in enumerate(a):

    #обязательно посещаем последнюю строку
    if i+1 == n:
        s += el
        continue

    if el > 0:
        s += el
        skip = False
        continue

    if el < 0 and skip: 
        continue
        

    if el < 0:
        #перепрыгиваем
        if a[i+1] > 0:
            continue
        elif k > 0:
            k += -1
            skip = True
        else:
            s += el

print(s)
```
