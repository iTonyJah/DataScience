## 🟢 4. Индексация и срезы.

Индексация в Pandas позволяет подступаться к строкам и столбцам** `DataFrame` или `Series`.

**Основные способы:**

* `[]` — простая индексация по имени столбца,
* `.loc[]` — доступ по **меткам** (именам),
* `.iloc[]` — доступ по **позициям** (номерам строк/столбцов).

С **индексацией столбцов** мы уже имели дело:

```python
# Один столбец возвращает Series
data['Age'].head()
```

```python
# Несколько столбцов возвращает DataFrame
data[['Name', 'Age', 'Fare']].head()
```

```python
# Можно обращаться к столбцу по его номеру
data.columns[8]
```

**Доступ к строкам:**

```python
# По индексу (позиции)
data.iloc[0]     # первая строка
```

```python
# По метке индекса (если переопределён, например, Name)
data.loc[0]      # тоже первая строка в текущем индексе
```

**Срезы строк** работают по аналогичному NumPy принципу: `.iloc` использует **позиции** (начало включительно, конец не включительно)

```python
# Первые 5 строк
data.iloc[:5]
```

```python
# Строки с 10 по 14
data.iloc[10:15]
```

Можно сделать срез одновременно и по строкам, и по столбцам - в этом тоже помогает `iloc`:

```python
# Возраст и пол первых 5 пассажиров
data.loc[:4, ['Name', 'Sex', 'Age']]
```

```python
# Первые 3 строки и первые 2 столбца
data.iloc[:3, :2]
```

Также нередко приходится выполнять **фильтрацию**, здесь главное -- правильно записать условие. Принцип почти тот же самый, что и в обычных `if`:

* каждое *атомарное условие* возвращает `True` или `False`;
* условие может быть *комплексным* -- т.е. состоять из нескольких *атомарных условий*, которые соединяются через "и", "или"; здесь важное отличие от `if`: **"и" записывается как `&`, "или"  как `|`**;
* чтобы избежать ошибок и неточностей, лучше обрамлять *атомарные условия* в отдельные скобки.

```python
# можно отфильтровать по любому условию
data[(data['Pclass'] == 3) | (data['Sex'].isin(data['Name'].values[:10]))].shape
```

```python
# пример полезного применения:
# посчитаем средний тариф для женщин и мужчин
data_male = data[data['Sex'] == 'male']
data_female = data[data['Sex'] == 'female']
print('Средний возраст женщин:', data_female['Age'].mean())
print('Средний возраст мужчин:', data_male['Age'].mean())
```

## 🟢 5. Изменение структуры и создание новых столбцов.

В предыдущих разделах мы познакомились с методами и атрибутами, которые позволяют просматривать и анализировать данные, не изменяя саму структуру/состав таблицы.

В этом разделе рассмотрим инструменты, которые могут вносить изменения в данные — например, добавлять, удалять или преобразовывать столбцы и строки.

Первым делом, давайте **переименуем** для удобства столбцы?

```python
# в этом поможет метод rename,
# не забываем сделать пересохранение!
# тут есть inplace=True !
data = data.rename(columns={
    'PassengerId': 'ID',
    'Survived': 'Выживаемость',
    'Pclass': 'Класс',
    'Name': 'Имя',
    'Sex': 'Пол',
    'Age': 'Возраст',
    'SibSp': 'Число_родств_1',
    'Parch': 'Число_родств_2',
    'Ticket': 'Билет',
    'Fare': 'Тариф',
    'Cabin': 'Каюта',
    'Embarked': 'Порт'
})
data.head()
```

Неинформативные или ненужные столбцы можно **удалять**:

```python
# в этом поможет метод drop
# главное - указывайте axis=1, чтобы удалить столбец
# (axis=0) отвечает за удаление строк
data = data.drop('ID', axis=1)
# data.drop('ID', axis=1, inplace=True) # если не хотите пересоздавать копию!
data.head()
```

А можно и **создавать новые**:

```python
data['Суммарное_число_родств'] = data['Число_родств_1'] + data['Число_родств_2']
```

Можно также применять функции, в т.ч. кастомные к столбцу/столбцам:

```python
# Создаем новую колонку с возрастной группой
data['Возрастная_группа'] = data['Возраст'].apply(lambda x: (x // 10) * 10)
data[['Возраст', 'Возрастная_группа']].head()
```

```python
# альтернативная и более универсальна форма записи для позапредыдущего примера -
# если хотим создать новый столбец из нескольких других
data['Суммарное_число_родств'] = data.apply(
    lambda row: row['Число_родств_1'] + row['Число_родств_2'],
    axis=1
)
```

Ещё иногда в данных встречаются **дубликаты**, что может мешать исследованиям:

```python
# метод drop_duplicates убирает дубликаты из данных
# тут есть inplace=True !
print(data.shape)
data = data.drop_duplicates()
print(data.shape)
```

```python
# можно убирать дубли по конкретным столбцам
print(data.shape)
tmp = data.drop_duplicates(['Пол', 'Возраст'])
print(tmp.shape)
```

```python
# заполним пропуски: для возраста - медианой
data['Возраст'].fillna(data['Возраст'].median(), inplace=True)

# для кабины - модой
data['Каюта'].fillna(data['Каюта'].mode()[0], inplace=True)
```

Пример из предыдущего раздела про расчёт среднего возраста в разрезе пола можно было реализовать проще и лаконичнее, используя **метод группировки** - вот как это бы выглядело:

```python
data.groupby(['Пол', 'Класс'])[['Возраст', 'Тариф']].mean()
```

```python
data.groupby('Пол')['Возраст'].mean()
```

```python
# reset_index() сделает отдельный числовой индекс,
# это может быть полезным при объединении таблиц
data.groupby('Пол')['Возраст'].mean().reset_index()
```

Ну и наконец последняя на сегодня (но не по важности:)) группа методов сохраняет преобразованный датафрейм в отдельный csv или excel файл:

```python
# параметр index=False откинет лишний столбец индекса
data.to_csv('titanic_preprocessed.csv', index=False)
data.to_excel('titanic_preprocessed.xlsx', index=False)
```
