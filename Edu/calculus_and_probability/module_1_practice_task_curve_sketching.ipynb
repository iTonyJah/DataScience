Исследуемая функция: 
$y = \frac{x^3}{2(x+5)^2}
$

```python
from sympy import plot, Symbol
x = Symbol("x") 
f = (x**3)/(2*(x+5)**2)
```

## Область определения

```python
from sympy import Symbol, S
from sympy.calculus.util import continuous_domain 
x = Symbol("x") 
continuous_domain(f, x, S.Reals)
```

```python
f
```

## Область значений функции

```python
from sympy import Symbol, S
from sympy.calculus.util import function_range

function_range(f, x, S.Reals)
```

```python
f
```

## Нахождение точек пересечения с осями координат

```python
from sympy import solveset, Eq
solveset(Eq(f, 0), x)
```

```python
f.subs(x, 0)
```

## Исследование функции на чётность и нечётность

$y(-x) = \frac{-x^3}{2(-x+5)^2}$

функция ни чётная, ни нечётная

## Нахождение производной

```python
f
```

```python
f.diff(x)
```

найдём точки, в которых она может менять знак

```python
y = f.diff(x)
solveset(Eq(y, 0), x)
```

```python
y
```

```python
solveset(Eq(y,0),1)
```

```python
scale = 20
plot(y, (x, -scale, scale), xlim=(-scale, scale), ylim=(-scale, scale))
```

## Нахождение второй производной

```python
f.diff(x,2)
```

```python
y = f.diff(x,2)
solveset(Eq(y, 0), x)
```
