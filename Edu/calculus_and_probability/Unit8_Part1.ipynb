```python
def f(x, y):
    return (1 - x) ** 2 + 100 * (y - x ** 2) ** 2
def grad(x, y):
    dx = 2 * x - 2 + 100 * (-4 * y * x + 4 * x ** 3)
    dy = 100 * (2 * y - 2 * x ** 2)
    return (dx, dy)
```

```python
x0 = (2, 2)
gamma = 0.001
x_cur = x0

vals = []
coords = []
i = 0
while True:
    x_new = (x_cur[0] - gamma * grad(*x_cur)[0],
            x_cur[1] - gamma * grad(*x_cur)[1])
    if i > 50000:
        break
    x_cur = x_new
    vals.append(f(*x_cur))
    coords.append(x_cur)
    i += 1
    if i % 10000 == 0:
        print(f"iter={i}; x=({x_cur[0]:.4f}, {x_cur[1]:.4f});"
          f" f(x)={f(*x_cur):.4f}; grad f(x)=({grad(*x_cur)[0]:.4f}, {grad(*x_cur)[1]:.4f})")
    
print(f"iter={i}; x=({x_cur[0]}, {x_cur[1]});"
        f" f(x)={f(*x_cur)}; grad f(x)=({grad(*x_cur)[0]}, {grad(*x_cur)[1]})")
```

```python
from scipy import optimize
```

```python
optimize.minimize(lambda x: f(*x), x0=(2, 2)) 
```
