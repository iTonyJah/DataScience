```python
import sympy as sp

# Определяем переменные
x, y = sp.symbols('x y')

# Задаем функцию
f = x**2 + x*y - y + y**3

# Находим частные производные
f_x = sp.diff(f, x)
f_y = sp.diff(f, y)

print("Функция: f(x,y) =", f)
print("Частная производная по x: f_x =", f_x)
print("Частная производная по y: f_y =", f_y)

# Решаем систему уравнений
critical_points = sp.solve([f_x, f_y], [x, y])
print("\nКритические точки:")
for point in critical_points:
    print(f"x = {point[0]}, y = {point[1]}")

# Вторые производные для анализа
f_xx = sp.diff(f_x, x)
f_yy = sp.diff(f_y, y)
f_xy = sp.diff(f_x, y)

print(f"\nВторые производные:")
print(f"f_xx = {f_xx}")
print(f"f_yy = {f_yy}")
print(f"f_xy = {f_xy}")

# Определитель Гессе
H = f_xx * f_yy - f_xy**2
print(f"\nОпределитель Гессе: H = {H}")

# Анализ критических точек
print("\nАнализ критических точек:")
for point in critical_points:
    x_val, y_val = point
    H_val = H.subs({x: x_val, y: y_val})
    f_xx_val = f_xx.subs({x: x_val, y: y_val})
    f_val = f.subs({x: x_val, y: y_val})
    
    print(f"\nТочка ({x_val}, {y_val}):")
    print(f"  f(x,y) = {f_val}")
    print(f"  H = {H_val}")
    print(f"  f_xx = {f_xx_val}")
    
    if H_val > 0 and f_xx_val > 0:
        print("  → Локальный минимум")
        min_point = (x_val, y_val)
        min_value = f_val
    elif H_val > 0 and f_xx_val < 0:
        print("  → Локальный максимум")
    elif H_val < 0:
        print("  → Седловая точка")
    else:
        print("  → Требуется дополнительный анализ")

# Вычисляем численные значения
print(f"\nЧисленные значения:")
for point in critical_points:
    x_val, y_val = point
    x_num = float(x_val)
    y_num = float(y_val)
    f_num = float(f.subs({x: x_val, y: y_val}))
    print(f"Точка ({x_num:.3f}, {y_num:.3f}): f = {f_num:.6f}")

# Находим точку минимума и округляем координату y
min_y = float(min_point[1])
print(f"\nКоордината по оси ординат в точке минимума: {min_y:.6f}")
print(f"Округлённый ответ: {min_y:.2f}")
```

```python
import numpy as np
from scipy.optimize import minimize

def f(params):
    x, y = params
    return x**2 + x*y - y + y**3

# Начальное приближение
x0 = [0, 0]

# Минимизация
res = minimize(f, x0, method='BFGS')
x_min, y_min = res.x
f_min = res.fun

print(f"Точка минимума: x = {x_min:.6f}, y = {y_min:.6f}")
print(f"Значение функции в минимуме: {f_min:.6f}")
print(f"Координата по оси ординат (ответ): {y_min:.2f}")
```
