#### Методы для доступа к элементам по индексу

- loc[]: Используется для доступа к элементам по индексу (метке), которая может быть строкой, числом или другим объектом.
- iloc[]: Используется для доступа к элементам по целочисленной позиции (аналогично индексам в списках Python).

- В случае срезов по строковым индексам при использовании loc[] обе границы среза включены.
- В случае срезов по целочисленным позициям срезы работают как в стандартных списках Python: верхняя граница не включается.
- При использовании iloc[] для срезов всегда используются целочисленные позиции, независимо от кастомного индекса.

```python
s
```

```python
# Срез от 'b' до 'd' включительно
print(s.loc['b':'d'])
```

```python
# Срез по целочисленным позициям (от 1 до 3, не включая 3)
print(s.iloc[1:3])
```

```python
# Неявный срез по строковому индексу
print(s['b':'d'])  # Это аналогично .loc['b':'d']
```

```python
# Неявный срез по числовым позициям
print(s[1:3])  # Это аналогично .iloc[1:3]
```

| Операция          | `.loc[]`                         | `.iloc[]`                    |
|-------------------|----------------------------------|------------------------------|
| Доступ по         | Индексу (именному)               | Позиции (числовой)           |
| Тип индекса       | Строки, числа, другие объекты    | Только целые числа (позиции) |
| Включение верхней границы в срезе | Включена                   | Не включена                  |
| Пример            | `s.loc['b']`                     | `s.iloc[1]`                  |
| Пример среза      | `s.loc['b':'d']` (включает 'd')  | `s.iloc[1:4]` (не включает 4)|

#### DataFrame
`DataFrame` — это двумерная табличная структура данных, которая представляет собой коллекцию `Series`, каждая из которых является столбцом. 

`DataFrame` можно рассматривать как аналог таблицы в базах данных или как электронную таблицу Excel.

##### Основные характеристики `DataFrame`:
- **Двумерная структура**: Данные представлены в виде строк и столбцов.
- **Гибкость типов данных**: Разные столбцы могут иметь разные типы данных (например, числа в одном столбце и строки в другом).
- **Индексация**: Строки и столбцы имеют индексы, которые можно использовать для доступа к данным.
- **Поддержка операций с табличными данными**: Pandas предоставляет инструменты для агрегации, сортировки, фильтрации и работы с отсутствующими значениями.

#### Оси

1. **Ось 0 (Axis 0)** — это строки (индексация по строкам).
   - Эта ось идет сверху вниз.
   - Многие операции по умолчанию выполняются вдоль оси 0, например, агрегационные функции вроде `sum()`, `mean()`, и т.д., которые по умолчанию вычисляются для каждого столбца, объединяя строки.

2. **Ось 1 (Axis 1)** — это столбцы (индексация по столбцам).
   - Эта ось идет слева направо.
   - При выполнении операций вдоль оси 1 агрегируются столбцы. Например, функция `sum(axis=1)` будет суммировать значения вдоль строк (по каждому столбцу).

<img src="axes.png" style="width: 400px;"/>

```python
data = {
    'A': [1, 2, 5],
    'B': [4, 3, 7],
    'C': [3, 6, 9]
}

df = pd.DataFrame(data)
df = df.transpose()
df
```

```python
df.axes
```

```python
df.sum()
```

```python
df.sum(axis=1)
```

### Примеры методов, использующих оси:

- `df.sum(axis=0)` — по умолчанию суммирует значения по оси 0 (по строкам для каждого столбца).
- `df.mean(axis=1)` — вычисляет среднее по строкам (агрегация столбцов).
- `df.drop(axis=1)` — удаляет столбец при использовании `axis=1`.
