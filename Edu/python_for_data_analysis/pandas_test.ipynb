## 5/10   Создание и преобразование столбцов с помощью функций

```python
import pandas as pd

def get_experience(arg):
    year_words = ['лет', 'года', 'год',]
    month_words = ['месяца', 'месяцев', 'месяц',]
    words = arg.split()
    experience_months = 0
    
    for i, word in enumerate(words):
        if word in year_words:
            experience_months += int(words[i-1]) * 12
        elif word in month_words:
            experience_months += int(words[i-1])
            
    return experience_months

```

```python
hh_df = pd.DataFrame({
    'experience': ['Опыт работы 8 лет 3 месяца', 
                   'Опыт работы 3 года 5 месяцев',
                   'Опыт работы 1 год 9 месяцев',
                   'Опыт работы 3 месяца',
                   'Опыт работы 6 лет',
                   'Опыт работы 1 год 1 месяц',
                  ],})

hh_df['exp_months'] = hh_df['experience'].apply(get_experience)
hh_df
```

##### Задание 6.3 объединить все файлы csv в папке, удалить дубликаты, вернуть DataFrame 



```python
import pandas as pd
import os

def concat_user_files(path):
    data = pd.DataFrame()
    file_names = os.listdir(path)
    file_names.sort()
    for file in file_names:
        tmp_data = pd.read_csv(path + '/' + file)
        data = pd.concat([data, tmp_data], axis=0, ignore_index=True)
    data = data.drop_duplicates()
    return data
```

Для решения задач нам понадобится выделить из признака title год выпуска фильма. Для этого напишем функцию get_year_release(arg).

```python
#библиотека для регулярных выражений
import re 
def get_year_release(arg):
    #находим все слова по шаблону "(DDDD)"
    candidates = re.findall(r'\(\d{4}\)', arg) 
    # проверяем число вхождений
    if len(candidates) > 0:
        #если число вхождений больше 0,
	#очищаем строку от знаков "(" и ")"
        year = candidates[0].replace('(', '')
        year = year.replace(')', '')
        return int(year)
    else:
        #если год не указан, возвращаем None
        return None
```
