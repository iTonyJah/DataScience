```python
import pandas as pd


def create_companyDF(income, expenses, years):
    df = pd.DataFrame({
    'Income': income,
    'Expenses': expenses,
    })
    df.index = years
    return df


def get_profit(df, year):
    
    if year in df.index:
        profit = df.loc[year, 'Income'] - df.loc[year, 'Expenses']
    else:
        profit = None
    
    return profit
    
```

## 5/10   Создание и преобразование столбцов с помощью функций

```python
import pandas as pd

def get_experience(arg):
    year_words = ['лет', 'года', 'год',]
    month_words = ['месяца', 'месяцев', 'месяц',]
    words = arg.split()
    experience_months = 0
    
    for i, word in enumerate(words):
        if word in year_words:
            experience_months += int(words[i-1]) * 12
        elif word in month_words:
            experience_months += int(words[i-1])
            
    return experience_months

```

```python
hh_df = pd.DataFrame({
    'experience': ['Опыт работы 8 лет 3 месяца', 
                   'Опыт работы 3 года 5 месяцев',
                   'Опыт работы 1 год 9 месяцев',
                   'Опыт работы 3 месяца',
                   'Опыт работы 6 лет',
                   'Опыт работы 1 год 1 месяц',
                  ],})

hh_df['exp_months'] = hh_df['experience'].apply(get_experience)
hh_df
```
