```python
def brackets(line):
    from collections import deque
    lq = deque('')
    for char in line:
        if char == '(':
            lq.append(char)
        elif char == ')':
            try:
                lq.pop()
            except:
                return False               
                
    return len(lq) == 0
    
    
```

```python
print(brackets("(()())"))
# True
print(brackets(""))
# True
print(brackets("(()()))"))
# False
```

```python
def flatten(lst):
    result = []
    for elem in lst:
        if type(elem) is list:
            result += flatten(elem)
        else:
            result.append(elem)

    return result
```

```python
tasks = [
            (36871, 'office', False),
            (40690, 'office', False),
            (35364, 'voltage', False),
            (41667, 'voltage', True),
            (33850, 'office', False),
        ]
```

```python
def task_manager(tasks):
    from collections import deque, defaultdict
    
    task_dict = defaultdict(deque)
    
    for task in tasks:
        if task[-1]:
            task_dict[task[1]].appendleft(task[0])
        else:
            task_dict[task[1]].append(task[0])
    
    return task_dict
```

```python
print(task_manager(tasks))
```

```python
print(task_manager(tasks))
# defaultdict(, {'voltage': deque([41667, 35364]),
# 'office': deque([36871, 40690, 33850])})
```
