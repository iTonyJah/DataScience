```python
def brackets(line):
    from collections import deque
    lq = deque('')
    for char in line:
        if char == '(':
            lq.append(char)
        elif char == ')':
            try:
                lq.pop()
            except:
                return False               
                
    return len(lq) == 0
    
    
```

```python
print(brackets("(()())"))
# True
print(brackets(""))
# True
print(brackets("(()()))"))
# False
```

```python
def flatten(lst):
    result = []
    for elem in lst:
        if type(elem) is list:
            result += flatten(elem)
        else:
            result.append(elem)

    return result
```
