```python
def check(temps):
    from collections import OrderedDict
    ordered_temps = OrderedDict(sorted(temps, key=lambda x: x[1], reverse=True))
    print(ordered_temps)

temps = [('2000', -4.4), ('2001', -2.5), ('2002', -4.4), ('2003', -9.5)]
check(temps)
```

```python
def brackets(line):
    from collections import deque
    lq = deque('')
    for char in line:
        if char == '(':
            lq.append(char)
        elif char == ')':
            try:
                lq.pop()
            except:
                return False               
                
    return len(lq) == 0
    
    
```

```python
print(brackets("(()())"))
# True
print(brackets(""))
# True
print(brackets("(()()))"))
# False
```
