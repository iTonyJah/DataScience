```python
import pandas as pd
import random
import numpy as np

# Списки имен и фамилий
female_names = ["Мария", "Елена", "Анна", "Ольга", "Татьяна", "Оксана", "Олеся", "Наталия", "Виктория", "Галина"]
male_names = ["Иван", "Сергей", "Алексей", "Дмитрий", "Андрей", "Евгений", "Петр", "Станислав", "Олег", "Кирилл"]
female_surnames = ["Петрова", "Попова", "Морозова", "Лебедева", "Кузнецова", "Смирнова", "Майорова", "Петухова", "Григорьева", "Добрынина"]
male_surnames = ["Иванов", "Смирнов", "Соколов", "Кузнецов", "Васильев", "Петров", "Горшков", "Борисов", "Пряников", "Федотов"]

diagnoses = ["Здоров", "Гипертония", "Диабет", "Ожирение", "Сердечная недостаточность"]

def generate_patient_data(n=1000):
    data = {
        "ID_пациента": range(1, n + 1),
        "ФИО": [],
        "Возраст": [random.choice([random.randint(18, 90), None]) for _ in range(n)],
        "Пол": [],
        "Рост_см": [random.choice([random.randint(150, 200), None]) for _ in range(n)],
        "Вес_кг": [random.choice([random.randint(50, 150), None]) for _ in range(n)],
        "Систолическое_АД": [random.choice([random.randint(90, 180), None]) for _ in range(n)],
        "Диастолическое_АД": [random.choice([random.randint(60, 120), None]) for _ in range(n)],
        "Уровень_холестерина": [random.choice([random.randint(3, 10) * 10, None]) for _ in range(n)],
        "Диагноз": [random.choice(diagnoses) for _ in range(n)],
    }

    for _ in range(n):
        gender = random.choice(["М", "Ж"])
        data["Пол"].append(gender)
        if gender == "М":
            name = random.choice(male_names)
            surname = random.choice(male_surnames)
        else:
            name = random.choice(female_names)
            surname = random.choice(female_surnames)
        data["ФИО"].append(f"{name} {surname}")

    # Добавим выбросы
    data["Рост_см"] = [x if random.random() > 0.02 else random.randint(50, 300) for x in data["Рост_см"]]
    data["Вес_кг"] = [x if random.random() > 0.02 else random.randint(30, 300) for x in data["Вес_кг"]]

    return pd.DataFrame(data)

# Генерация данных
patient_data = generate_patient_data(1000)

# Сохранение в CSV
file_path = "медицинские_данные.csv"
patient_data.to_csv(file_path, index=False, encoding="utf-8")
print(f"Данные сохранены в файл: {file_path}")
```
