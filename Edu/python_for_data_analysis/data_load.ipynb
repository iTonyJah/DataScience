В CSV-файлах могут использоваться разные символы в качестве разделителей:

1. **Запятая (,)**
   
2. **Точка с запятой (;)** – удобно если в данных запятая служит десятичным разделителем.

3. **Табуляция (`\t`)** – используется в формате TSV (Tab-Separated Values). Применяется, если данные содержат запятые и точки с запятой.

4. **Пробел ( )** – редко используется, но может встретиться, если данные не содержат пробелов внутри значений.

5. **Вертикальная черта (|)** – также известная как "pipe". Удобен для данных, где запятые и точки с запятой часто встречаются внутри значений.

6. **Двоеточие (:)** – встречается реже, но иногда применяется для упрощенных форматов данных.

Выбор разделителя зависит от контекста данных и специфики среды, где файл будет использоваться.

### xlsx файлы 

````python
import re
from zipfile import ZipFile
import pandas as pd

# Список для хранения DataFrame из каждого файла
dataframes = []

# Открытие ZIP-архива и обработка файлов
with ZipFile('data_zis_109_v20241014_xlsx.zip') as z:
    # Перебор файлов в архиве
    for file_info in z.infolist():
        print(file_info)
        # Отбираем нужные файлы
        if re.search(r'20(22|23)\.xlsx$', file_info.filename):
            with z.open(file_info.filename) as f:
                # Чтение файла в DataFrame и добавление в список
                df = pd.read_excel(f)
                dataframes.append(df)

# Объединение всех DataFrame
combined_df = pd.concat(dataframes, ignore_index=True)

combined_df.head()
````

**JSON** (JavaScript Object Notation) — это лёгкий текстовый формат обмена данными, который используется для передачи данных между сервером и клиентом. JSON позволяет структурировать данные в виде ключ-значение, что делает его удобным для хранения и обмена информацией в веб-приложениях и API.

### Структура JSON
JSON состоит из:
1. **Объектов** — это структуры данных, заключенные в фигурные скобки `{}`, где данные представлены в виде пар "ключ-значение". Например:

   ```json
   {
     "name": "Иван",
     "age": 30,
     "city": "Москва"
   }
   ```

2. **Массивов** — это списки значений, заключенные в квадратные скобки `[]`. Массивы могут содержать объекты, строки, числа и другие массивы. Например:

   ```json
   {
     "employees": [
       {"name": "Иван", "age": 30},
       {"name": "Анна", "age": 25},
       {"name": "Петр", "age": 35}
     ]
   }
   ```

3. **Простых значений** — это строки, числа, логические значения (`true` и `false`) и `null`.

### Применение JSON
JSON широко используется для:
- **Веб-API**: Передача данных между сервером и клиентом в формате JSON позволяет легко обмениваться структурированной информацией.
- **Хранение данных**: JSON-файлы часто используются для хранения настроек, конфигураций и других структурированных данных.

**XML** (Extensible Markup Language) — это язык разметки, предназначенный для хранения и передачи структурированных данных. XML-файлы широко используются для обмена информацией между различными системами, особенно в случаях, когда требуется передать сложные и иерархически организованные данные.

### Структура XML
XML-файл состоит из **тегов**, которые организуют данные в иерархическую структуру. Основные элементы XML:

1. **Теги** — элементы, заключённые в угловые скобки (`<tag>...</tag>`), которые обозначают начало и конец блока данных.
   
   ```xml
   <name>Иван</name>
   ```

2. **Атрибуты** — параметры внутри открывающего тега, которые добавляют свойства элементу.

   ```xml
   <employee id="123" department="HR">Иван</employee>
   ```

3. **Иерархия** — XML поддерживает вложенность, что позволяет создавать структуры данных с несколькими уровнями.

   ```xml
   <employees>
       <employee id="123" department="HR">
           <name>Иван</name>
           <age>30</age>
       </employee>
       <employee id="124" department="IT">
           <name>Анна</name>
           <age>25</age>
       </employee>
   </employees>
   ```

### Применение XML
XML часто используется для:

- **Обмена данными** между системами, так как легко читается и может быть проанализирован различными программами.
- **Конфигурационных файлов**: XML позволяет хранить настройки и параметры для приложений.
- **Хранения документов и данных** в иерархическом виде.
- **Веб-технологий**, например, в веб-сервисах (SOAP) и RSS-лентах.

````python
# URL API для получения данных о Нобелевских премиях
url = "https://api.nobelprize.org/2.1/nobelPrizes"

# Параметры запроса (можно указать год или категорию)
params = {
    "format": "json"
}

# Отправка запроса
response = requests.get(url, params=params)
data = response.json()
data
````

````python
data.keys()
````

````python
pd.DataFrame(data['nobelPrizes'])
````

````python
data['nobelPrizes'][2]
````

````python
prizes = []
for prize in data['nobelPrizes']:
    for laureate in prize.get('laureates', []):
        prizes.append({
            "year": prize['awardYear'],
            "category": prize['category']['en'],
            "laureate_name": laureate.get('knownName', {}).get('en', 'N/A'),  # обработка отсутствующего 'knownName'
            "motivation": laureate.get('motivation', {}).get('en', 'N/A')
        })
        
df = pd.DataFrame(prizes)

df.head()
````
