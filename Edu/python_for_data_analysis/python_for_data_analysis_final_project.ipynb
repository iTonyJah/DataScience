1. Начнем с простого - с признака **"Образование и ВУЗ"**. Его текущий формат это: **<Уровень образования год выпуска ВУЗ специальность...>**. Например:
* Высшее образование 2016 Московский авиационный институт (национальный исследовательский университет)...
* Неоконченное высшее образование 2000  Балтийская государственная академия рыбопромыслового флота…
Нас будет интересовать только уровень образования.

Создайте с помощью функции-преобразования новый признак **"Образование"**, который должен иметь 4 категории: "высшее", "неоконченное высшее", "среднее специальное" и "среднее".

Выполните преобразование, ответьте на контрольные вопросы и удалите признак "Образование и ВУЗ".

Совет: обратите внимание на структуру текста в столбце **"Образование и ВУЗ"**. Гарантируется, что текущий уровень образования соискателя всегда находится в первых 2ух слов и начинается с заглавной буквы. Воспользуйтесь этим.

*Совет: проверяйте полученные категории, например, с помощью метода unique()*


```python
# заведём функцию для обработки строки столбца таблицы, в дальнейшем будем обращаться к ней
def get_select(selection, splitby=' ', column_start=0, column_end=None, mapto=None):
    """
    Обрабатывает строку: разделяет её, извлекает поддиапазон "столбцов"
    и опционально сопоставляет результат со словарем.

    Args:
        selection: Входная строка с элементами.
        splitby: Разделитель. По умолчанию ' '.
        column_start: Начальный индекс поддиапазона для возвращения. По умолчанию 0.
        column_end: Конечный индекс. По умолчанию None.
        mapto (dict, optional): Словарь для сопоставления извлеченной строки. По умолчанию None.

    Returns:
        str: Извлеченная и объединенная подстрока или значение из словаря mapto.
    """
    selection_list = selection.split(splitby)
    select = ' '.join(selection_list[column_start:column_end])
    if mapto:
        select = mapto.get(select)
    return select

education_map = {
    'Высшее образование': 'высшее',
    'Неоконченное высшее': 'неоконченное высшее',
    'Среднее специальное': 'среднее специальное',
    'Среднее образование': 'среднее',
}

# так как удаляем колонку, то добавляем условие, чтобы при повторном запуске ячейки не вызывать ошибку
if 'Образование и ВУЗ' in df.columns:
    df['Образование'] = df['Образование и ВУЗ'].apply(get_select, column_start=0, column_end=2, splitby=' ', mapto = education_map)
    df = df.drop('Образование и ВУЗ', axis='columns')
df['Образование'].value_counts()
```

3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
# так как удаляем колонку, то добавляем условие, чтобы при повторном запуске ячейки не вызывать ошибку
if 'Опыт работы' in df.columns:
    df['Опыт работы'] = df['Опыт работы'].replace('Не указано', np.nan)
    
    experience = pd.DataFrame()
    
    # отделяем двойным пробелом текст формата "Опыт работы 16 лет 10 месяцев"
    experience['text'] = df['Опыт работы'][df['Опыт работы'].notna() == True].apply(get_select, splitby = '  ', column_end = 1)
    print('---Первоначальный текст---')
    display(experience.tail())
    
    # убираем "Опыт работы " из текста
    experience['text'] = experience['text'].apply(get_select, splitby = 'ы ', column_start = 1, column_end = 2)
    
    # добавим столбец с счётчиком проблеов
    experience['spaces'] = experience['text'].apply(lambda x: x.count(' '))
    
    # разобьём весь текст по пробелам
    splitted_text = experience['text'].str.split(expand=True)
    print('---Текст разбитый по столбцам---')
    display(splitted_text.tail())
    
    print("Уникальные значения типов времени: ", pd.concat([splitted_text[1], splitted_text[3]]).unique())
    
    year_name_list = ['лет', 'года', 'год',]
    month_name_list = ['месяцев', 'месяца', 'месяц',]
    
    splitted_text['first_factor'] = splitted_text[1].apply(lambda x: 12 if x in year_name_list else 1 if x in month_name_list else 0)
    splitted_text['second_factor'] = splitted_text[3].apply(lambda x: 12 if x in year_name_list else 1 if x in month_name_list else 0)
    # заменим все пропущенные значения на нули и преобразуем в числа числовые столбцы (0, 2)
    splitted_text = splitted_text.fillna(0)
    for column_index in [0, 2]:
        splitted_text[column_index] = splitted_text[column_index].astype(int)
    
    # переведём всё в месяцы и просуммируем
    splitted_text['first_product'] = splitted_text[0] * splitted_text['first_factor']
    splitted_text['second_product'] = splitted_text[2] * splitted_text['second_factor']
    splitted_text['total_months'] = splitted_text['first_product'] + splitted_text['second_product']
    
    # создадим требуемый столбец, None оставляем, остальное пока обнулим
    df['Опыт работы (месяц)'] = df['Опыт работы'].apply(lambda x: x if pd.isna(x) else 0)
    # добавим вычисленные месяца в непустые строки с опытом работы
    df.loc[df['Опыт работы (месяц)'].notna(), 'Опыт работы (месяц)'] = splitted_text['total_months']
    # Преобразуем в специальный целочисленный тип, который поддерживает NaN (pd.NA)
    df['Опыт работы (месяц)'] = df['Опыт работы (месяц)'].astype('Int64') # Обратите внимание на заглавную 'I'
    df = df.drop('Опыт работы', axis = 'columns')
print("Медианный опыт работы (в месяцах) равен ", df['Опыт работы (месяц)'].median())
print('---количество пропущенных значений---')
display(df['Опыт работы (месяц)'].isna().sum())
print('-'*19)
```

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
self_city = ['Москва', 'Санкт-Петербург',]
million_cities = [
    'Новосибирск', 'Екатеринбург','Нижний Новгород', 'Казань', 'Челябинск', 'Омск', \
    'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж','Волгоград',
]

# так как удаляем колонку, то добавляем условие, чтобы при повторном запуске ячейки не вызывать ошибку
if 'Город, переезд, командировки' in df.columns:
    df['Город'] = df['Город, переезд, командировки'].apply(get_select, splitby=' , ', column_start=0, column_end=1,)
    df['Город'] = df['Город'].apply(lambda x: x if x in self_city else 'город-миллионник' if x in million_cities else 'другие')
    
    # заведём временный признак для выделения строк с метро
    df['is_subway'] = df['Город, переезд, командировки'].apply(get_select, splitby=' , ', column_start=1, column_end=2,)
    df.is_subway = df.is_subway.str.contains('перее') == False
    
    # в зависимости от наличия "метро" по разному разобьём строки
    df['temp'] = df['Город, переезд, командировки'][df.is_subway == True].apply(get_select, splitby=' , ', column_start=2, column_end=3)
    df['temp2'] = df['Город, переезд, командировки'][df.is_subway == False].apply(get_select, splitby=' , ', column_start=1, column_end=2)
    df['temp'] = df['temp'].fillna(df['temp2'])
    df.temp2 = None
    # оставим только первые два слова
    df.temp = df.temp.apply(get_select, splitby=' ', column_start=0, column_end=2,)
    print("Варианты готовности к переезду", df.temp.unique())
    ready_for_relocation = {'готов к', 'готова к', 'хочу переехать'}
    df.temp = df.temp.apply(lambda x: True if x in ready_for_relocation else False)
    df['Готовность к переезду'] = df.temp
    df.temp = None

    # получим следующее разделение с готовностью к командировкам, в зависимости от наличия "метро"
    df['temp'] = df['Город, переезд, командировки'][df.is_subway == True].apply(get_select, splitby=' , ', column_start=3,)
    df['temp2'] = df['Город, переезд, командировки'][df.is_subway == False].apply(get_select, splitby=' , ', column_start=2)
    df['temp'] = df['temp'].fillna(df['temp2'])
    print('Варианты готовности к командировкам', df.temp.unique())
    # создадим множество вариантов _готовых_ к командировкам, остальные считаем, что не готовы
    # 'г' - считаю, что обрезанное "готов"
    willing_to_travel = {
        'готов к редким командировкам', 'готова к командировкам', 'готов к командировкам',
        'готова к редким командировкам', 'готов к команд', 'готов к', 'готов к командир', 'г',    
    }
    df['Готовность к командировкам'] = df.temp.apply(lambda x: True if x in willing_to_travel else False)
    df = df.drop(['is_subway', 'temp', 'temp2', 'Город, переезд, командировки'], axis='columns')

display(df['Город'].value_counts(normalize=True))

mask = (df['Готовность к переезду'] == True) & (df['Готовность к командировкам'] == True)
print('Готовы одновременно и к переездам, и к командировкам: ', df[mask].shape[0] / df.shape[0])
```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Перед вами таблица соответствия наименований иностранных валют в наших данных и их общепринятых сокращений, которые представлены в нашем файле с курсами валют. Теперь у нас есть вся необходимая информация для того, чтобы создать признак "ЗП (руб)" - заработная плата в рублях.

После ответа на контрольные вопросы удалите исходный столбец заработной платы "ЗП" и все промежуточные столбцы, если вы их создавали.

Итак, давайте обсудим возможный алгоритм преобразования: 
1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
fx = pd.read_csv('ExchangeRates.csv')
fx['date'] = pd.to_datetime(fx['date'], format='%d/%m/%y').dt.date

if 'ЗП' in df.columns:    
    df['amount'] = df['ЗП'].apply(get_select, column_end=1).astype(int)
    df['currency_input'] = df['ЗП'].apply(get_select, column_end=2, column_start=1)
    print('---виды валют ЗП---')
    display(df.currency_input.unique())
    currency_ISO_map = {
        'руб.': 'RUB', 
        'бел.руб.': 'BYN', 
        'KZT': 'KZT', 
        'EUR': 'EUR', 
        'USD': 'USD', 
        'грн.': 'UAH', 
        'сум': 'UZS', 
        'KGS': 'KGS', 
        'AZN': 'AZN',
    }
    df['currency'] = df.currency_input.apply(lambda x: currency_ISO_map.get(x))
    df = df.drop('currency_input', axis='columns')
    
    df['Обновление резюме'] = pd.to_datetime(df['Обновление резюме'], format='%d.%m.%Y %H:%M').dt.date
    df['date'] = df['Обновление резюме']
    
    merged = pd.merge(df, fx, on=['date', 'currency'], how='left')
    merged.loc[df.currency == 'RUB', ['close', 'proportion']] = 1
    
    df['ЗП (руб)'] = (merged.amount * merged.close / merged.proportion).astype(int)
    df = df.drop(['ЗП', 'amount', 'currency', 'date',], axis='columns')

# Чему равна желаемая медианная заработная плата соискателей в нашей таблице (в рублях)?
print('Желаемая медианная заработная плата: ', df['ЗП (руб)'].median())
```

```python
# 2. Построение сводной таблицы (pivot table)
# Агрегируем данные, используя медианное значение заработной платы
# Индексом (строками) будет 'Возраст', колонками - 'Образование'
pivot_table = df.pivot_table(
    values='ЗП (руб)',
    index='Возраст',
    columns='Образование',
    aggfunc='median'
)

print("Сводная таблица:")
print(pivot_table)

# 3. Построение тепловой карты на основе сводной таблицы с помощью Plotly Express
# Plotly Express может напрямую использовать DataFrame сводной таблицы
fig = px.imshow(pivot_table,
                labels=dict(x="Уровень образования", y="Возраст", color="Медианная желаемая зарплата"),
                x=pivot_table.columns,
                y=pivot_table.index,
                aspect="auto", # Автоматическое определение соотношения сторон
                title="Тепловая карта медианной желаемой заработной платы по возрасту и образованию",
               )

# Добавление текстовых аннотаций значений на тепловую карту (опционально)
fig.update_traces(textfont_size=12, selector=dict(type='heatmap'))

# Отображение графика
fig.show()
```

По **тепловой карте** хорошо видно, что медианная желаемая заработная плата интенсивней растёт с возрастом у людей с высшим образованием.

Опыт работы переведите из месяцев в года, чтобы признаки были в едином масштабе. Постройте на графике дополнительно прямую, проходящую через точки (0, 0) и (100, 100). Данная прямая соответствует значениям, когда опыт работы равен возрасту человека. Точки, лежащие на этой прямой и выше нее - аномалии в наших данных (опыт работы больше либо равен возрасту соискателя)

```python

# Преобразование опыта работы из месяцев в года ---
df['Опыт работы (год)'] = df['Опыт работы (месяц)'] / 12


# Построение диаграммы рассеяния с Plotly ---

# Создаем базовый график рассеяния
fig = px.scatter(df,
                 x='Возраст',
                 y='Опыт работы (год)',
                 title='Зависимость опыта работы от возраста соискателей',
                 labels={'Возраст': 'Возраст (года)', 'Опыт работы (год)': 'Опыт работы (года)'},
                 hover_data={'Опыт работы (месяц)': True, 'Опыт работы (год)': ':.1f'}
                )

# Добавляем линию, где Опыт == Возраст
# Точки на этой линии или выше нее - аномалии
fig.add_shape(
    type='line',
    x0=0, y0=0, x1=100, y1=100, # Координаты от (0,0) до (100,100)
    line=dict(
        color='Red',
        width=2,
        dash='dash',
    ),
    name='Линия аномалий (Опыт >= Возраст)'
)

# Обновляем макет для лучшей читаемости и отображения легенды линии
fig.update_layout(
    xaxis_range=[0, df['Возраст'].max() + 5],
    yaxis_range=[0, df['Опыт работы (год)'].max() + 5],
    showlegend=True
)

# Отображаем график
fig.show()
```

На графике хорошо видны **выбросы**, точки лежащие выше линии (опыт работы больше возраста) и на самой линии. Ещё три точки лежат близко к линии (опыт работы с 5 лет).
