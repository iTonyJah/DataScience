Для анализа взаимодействия факторов factor_3 и factor_4 с помощью симуляции MVT (Multivariate Testing) и определения p-value для взаимодействия, мы можем использовать логистическую регрессию, добавив в неё член взаимодействия.

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from scipy import stats
```

```python
# Загрузка данных
df = pd.read_csv('mvt_experiment_dataset.csv')

# Логистическая регрессия с взаимодействием факторов
model = smf.logit('conversion ~ factor_3_Reviews_Placement * factor_4_Guarantee_Badge', data=df).fit(disp=False)

# Поиск p-value для членов взаимодействия
interaction_terms = [term for term in model.pvalues.index if ':' in term or '*' in term]
```

# Практическое задание 2.3. «Оптимизация размера выборки»

## Содержание задания

Рассчитать минимальный размер выборки на группу для MVT-теста с 12 группами, чтобы детектировать эффект 5% с мощностью 80% и уровнем значимости 5%.

## Параметры расчета
- **Количество групп**: 12
- **Минимальный детектируемый эффект**: 5% (0.05)
- **Мощность теста**: 80% (β = 0.2)
- **Уровень значимости**: 5% (α = 0.05)
- **Тип теста**: Двусторонний

## Методология
Для решения задачи нам нужно рассчитать минимальный размер выборки на группу для MVT-теста с 12 группами, чтобы детектировать эффект 5% с мощностью 80% и уровнем значимости 5%.
Мы будем использовать формулу для расчета размера выборки в A/B-тестах (или MVT) для пропорций (конверсии). 
### Используемая формула
Для расчета размера выборки на группу используется стандартная формула для сравнения пропорций:


$n = \frac{(Z_{1-α/2} + Z_{1-β})² × (p₁(1-p₁) + p₂(1-p₂))}{(p₁ - p₂)²}$

где:
- $p₁$ - базовая конверсия (контроль)
- $p₂ = p₁ + эффект$ - конверсия с эффектом
- $Z_{1-\frac{α}{2}}$ - Z-score для уровня значимости
- $Z_{1-β}$ - Z-score для мощности

### Расчетные значения
- **$Z_{1-α/2}$** (α=0.05, двусторонний): 1.9695
- **$Z_{1-β}$** (β=0.20): 0.8416

## Результаты расчета

Из данных определяется базовая конверсия, затем рассчитывается необходимый размер выборки:

Базовая конверсия (p₁): 0.14

Целевая конверсия (p₂): 0.19

Расчетный размер выборки на группу: 862

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('mvt_experiment_dataset.csv')

# Базовая конверсия
p1 = df['conversion'].mean()
effect = 0.05
p2 = p1 + effect

alpha = 0.05
beta = 0.20

z_alpha = stats.norm.ppf(1 - alpha/2)
z_beta = stats.norm.ppf(1 - beta)
print('z_alpha, z_beta', z_alpha, z_beta)

n = ( (z_alpha + z_beta)**2 * (p1*(1-p1) + p2*(1-p2)) ) / ( (p1 - p2)**2 )
n_per_group = int(np.ceil(n))

print(f"Baseline conversion: {p1:.3f}")
print(f"Target conversion: {p2:.3f}")
print(f"Required sample size per group: {n_per_group}")
```

# Расчет минимальной длительности теста (Практическое задание 2.5)

## Условие задачи

Необходимо рассчитать минимальную длительность A/B/n теста в днях при следующих вводных данных:

*   **Ежедневный трафик:** 5000 пользователей
*   **Количество групп в MVT (N):** 16
*   **Минимальный размер группы (S_min):** 1000 пользователей

*Примечание: округление длительности осуществляется до целых единиц времени в большую сторону.*

---

## Решение

### 1. Расчет необходимого общего объема выборки

Сначала определим общее минимальное количество пользователей, которое должно пройти через тест, чтобы в каждой из 16 групп набралось минимум по 1000 человек.

Используем формулу:
$$
S_{total} = N \times S_{min}
$$

Подставляем значения:
$$
S_{total} = 16 \times 1000 = 16000 \text{ пользователей}
$$

### 2. Определение расчетной длительности в днях

Далее рассчитываем количество дней, которое потребуется для набора 16 000 пользователей при ежедневном трафике в 5000 пользователей.

Используем формулу:
$$
D_{calc} = \frac{S_{total}}{\text{Ежедневный трафик}}
$$

Подставляем значения:
$$
D_{calc} = \frac{16000}{5000} = 3.2 \text{ дня}
$$

### 3. Округление результата

Согласно условиям проведения A/B-тестов, длительность всегда округляется до целого числа в большую сторону (функция "потолок" $\lceil x \rceil$), так как тест нельзя остановить в середине дня.

$$
D_{final} = \lceil 3.2 \rceil = 4 \text{ дня}
$$

---

## **Ответ:**

Минимальная длительность теста составляет **4 дня**.
