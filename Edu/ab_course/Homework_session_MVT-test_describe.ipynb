# MVT-тестирование (Multi-Variable Testing)

MVT-тестирование - это тестирование нескольких факторов (A, B, C) с несколькими вариациями (A1, A2, B1, B2...) одновременно.

**Цель**: определить не только лучшие комбинации, но и взаимодействие факторов.

**Пример**: Заголовок (A), Изображение (B), Цвет кнопки (C).

**Продолжительность теста**: Минимальная длительность (дни) = ```ceil```( (Минимальный размер группы × Количество комбинаций) / Ежедневный трафик )).

## 📋 Часть 1: Ключевые шаги планирования и проведения MVT-теста
Проведение мультивариантного теста — это структурированный процесс, который можно разбить на три основные фазы.

### 🎯 Фаза 1: Подготовка и планирование
На этом этапе закладывается фундамент для успешного эксперимента.

1. **Определение проблемы и цели**: Анализируем данные (например, из Google Analytics, тепловых карт) и обратную связь от пользователей, чтобы найти «слабое звено» — страницу с низкой конверсией или высоким процентом отказов. Чётко формулируем бизнес-цель: «увеличить конверсию в заказ на продуктной странице на 15%».
2. **Формулировка гипотезы**: На основе выявленной проблемы выдвигаем проверяемое предположение. Качественная гипотеза строится по схеме: «Если мы изменим (*элементы A, B, C*) на (*A1, A2, B1, B2...*), то (*метрика*) улучшится, потому что (*обоснование*)».
3. **Выбор элементов и создание вариаций**: Определяем, какие именно элементы на странице (например, заголовок, основное изображение, текст CTA-кнопки (Call To Action, кнопка призыва к действию)) мы будем тестировать. Подготовка для каждого элемента 2-3 варианта. Критически важно не смешивать в одном тесте независимые факторы, такие как макет страницы и её контент.
4. **Расчет количества комбинаций и выбор метода**: Количество версий страницы равно произведению вариантов каждого элемента (2 заголовка × 3 изображения × 2 CTA-кнопки = 12 комбинаций). Исходя из этого числа и доступного трафика, выбираем метод:
    * **Полнофакторный (Full Factorial)**: Тестирует все комбинации, давая максимально точные данные об эффектах взаимодействия, но требует очень много трафика и времени.
    * **Дробный (Fractional Factorial)**: Тестирует только часть комбинаций, статистически вычисляя остальные. Требует меньше трафика, но даёт менее детальные insights.

### 🚀 Фаза 2: Запуск и выполнение

5. **Определение размера выборки и длительности теста**: Определяем текущую конверсию (Baseline Conversion Rate), минимальный ожидаемый эффект (Minimum Detectable Effect) и желаемый уровень достоверности (Statistical Significance, обычно 95%). Используем определённый калькулятор для расчёта метрик. Тест должен длиться до тех пор, пока каждая комбинация не наберет рассчитанное количество просмотров. **Важно запускать все варианты одновременно**, чтобы нивелировать влияние внешних факторов (день недели, сезонность).
6. **Настройка и запуск**: С помощью специальных платформ (VWO, Optimizely, Adobe Target) создаём эксперимент, загружаем все варианты контента и настраиваем равномерное распределение трафика. Перед запуском обязательно проверяем техническую исправность всех элементов (кликабельность кнопок, работа форм).

### 📊 Фаза 3: Анализ результатов и выводы

7. **Сбор данных и оценка статистической значимости**: После завершения теста используем тот же калькулятор для проверки статистической значимости результатов. Это покажет, являются ли наблюдаемые различия достоверными, а не случайными.
8. **Интерпретация взаимодействий и выбор победителя**: Анализируем, не только какая комбинация показала лучший результат, но и как отдельные элементы влияли друг на друга. Сравниваем выигрышную комбинацию с оригинальной (контрольной) версией. Принимаем решение о внедрении победившей версии, запуске нового раунда тестов для уточнения или возврате к контрольному варианту.

# ❓ Часть 2: Ответы на подвопросы

## 1. В чем основное отличие MVT от A/B-тестирования?
Ключевое отличие — в масштабе изменений и типе получаемых данных. A/B-тест (включая A/B/n) сравнивает две или несколько цельных версий страницы, часто с радикальными отличиями в дизайне или контенте. Он отвечает на вопрос: «Какая из этих страниц работает лучше?». MVT же тестирует множество комбинаций отдельных элементов (заголовков, изображений, кнопок) в рамках одной страницы. Он отвечает на вопросы: «Какая конкретная комбинация элементов работает лучше всего?» и «Как эти элементы взаимодействуют друг с другом?». MVT — это инструмент для «тонкой настройки», в то время как A/B-тест часто используется для проверки крупных гипотез.

## 2. Какие метрики обычно используются для оценки эффективности MVT?

Для оценки эффективности MVT (мультивариантного тестирования) используются стандартные метрики веб-аналитики, ориентированные на конверсию, вовлеченность и финансовые показатели: коэффициент конверсии (CR), средняя ценность конверсии (Average Value), доход на пользователя (ARPU), CTR, время на странице и глубина сессии для выявления лучшей комбинации элементов страницы, которая максимизирует бизнес-цели, например, продажи или регистрации. 

**Ключевые метрики эффективности MVT**:
* **Коэффициент конверсии (Conversion Rate, CR)**: Процент пользователей, совершивших целевое действие (покупка, подписка) в каждой из вариаций.
* **Средняя ценность конверсии (Average Order Value, AOV/ARPU)**: Средний чек или доход, приносимый пользователем в каждой комбинации, критически важен для eCommerce.
* **Доход (Revenue)**: Общий доход, сгенерированный каждой комбинацией элементов.
* **CTR (Click-Through Rate)**: Эффективность конкретных кнопок, заголовков или изображений.
* **Время на странице/глубина сессии (Time on Page/Session Depth)**: Показывает, насколько вариация удерживает внимание пользова
теля.
* **Показатель отказов (Bounce Rate)**: Снижение отказов при переходе на новую страницу указывает на повышение релевантности.
* **Значимость (Statistical Significance)**: Вычисляется, чтобы убедиться, что результаты не случайны. 

## 3. Как определить количество групп (комбинаций) в MVT?
Количество тестируемых групп-комбинаций рассчитывается по формуле:

[Кол-во вариантов элемента A] × [Кол-во вариантов элемента B] × [Кол-во вариантов элемента C] … = [Общее кол-во комбинаций].

**Пример**: Если тестируем 2 варианта заголовка, 3 варианта изображения и 2 варианта текста на кнопке, то получится 2 × 3 × 2 = 12 уникальных комбинаций (версий страницы).

**Важное ограничение**: С увеличением числа комбинаций резко возрастает требуемый объём трафика и время теста. Рекомендуется начинать с 2-3 наиболее важных элементов.

**Пример с картинкой:** 

2 (Product image) x 2 (CTA button color) x 2 (Product headline) = 8

## 4. Какие риски связаны с проведением MVT?

* **Недостаточный трафик** — главный риск. Для полнофакторного MVT требуется очень высокая посещаемость (рекомендуется от 100 000+ уникальных посетителей в месяц). При недостатке трафика тест не достигнет статистической значимости за разумное время.
* **Ложные срабатывания** (False Positives). Чем больше комбинаций тестируется, тем выше вероятность случайно получить «выигрышный» вариант из-за шума в данных. Необходима перепроверка результатов.
* **Сложность настройки и анализа**. MVT требует больше времени и экспертизы для корректной настройки, а интерпретация результатов (особенно эффектов взаимодействия) сложнее, чем в A/B-тесте.
* **Неверный выбор элементов**. Включение в тест малозначимых элементов (например, изменение футера сайта) увеличивает сложность, не давая существенной пользы.
* **Данные ненормальные/сильно скошены** (например, доход на пользователя). Стандартный **ANOVA** может быть неэффективен.
* * Решение: Применить **непараметрические методы** (тест Манна-Уитни для медиан) или **бутстреп** для оценки доверительных интервалов. Или использовать **обобщенные линейные модели (GLM)**, предназначенные для конкретных типов данных (например, Гамма-распределение для сильно скошенных непрерывных данных).

## 5. Как интерпретировать взаимодействие факторов в MVT?


Взаимодействие факторов — это ключевое преимущество MVT, показывающее, что эффект от изменения одного элемента **зависит** от того, какой вариант другого элемента представлен на странице.

**Пример интерпретации**: Тестируем заголовок («Скидка» / «Выгода») и цвет кнопки (красный / зелёный).

* **Если нет взаимодействия**: Допустим, «Скидка» всегда лучше «Выгоды» на 5%, а красная кнопка всегда лучше зелёной на 3%. Их эффекты просто суммируются.
* **Если есть взаимодействие**: Может оказаться, что комбинация «Скидка + зелёная кнопка» даёт аномально высокий результат, хотя по отдельности зелёная кнопка проигрывала. Это означает, что эти два фактора **синергируют** — вместе они работают лучше, чем можно было ожидать от их отдельных эффектов. И наоборот, факторы могут **конфликтовать**, ухудшая общий результат.

Для выявления и оценки силы этих взаимодействий в MVT используется **дисперсионный анализ (ANOVA)**.

**Практические шаги для интерпретации**:
* **Анализ таблицы результатов**: Смотрим на матрицу результатов. Если ячейки не просто суммируются, а показывают всплеск или падение, это взаимодействие.
* **Статистическая значимость**: Проверяем значимо ли это взаимодействие, а не просто случайность.
* **Визуализация**: Стром тепловые карты или графики, чтобы увидеть, как изменение одного фактора (например, цвет кнопки) меняет эффективность другого (например, размер изображения).
