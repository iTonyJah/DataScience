```python
# test_manual.py — быстрая проверка RPC
import zerorpc

# Подключаемся к запущенному сервису
client = zerorpc.Client()
client.connect("tcp://127.0.0.1:8000")

# 1. Проверяем health (если есть эндпоинт)
# 2. Загружаем данные (prepare)
resp = client.prepare(
    user_path="data/train/user.csv",
    shift_path="data/train/shift.csv", 
    event_path="data/train/event.csv"
)
print("✅ prepare:", resp)

# 3. Проверяем готовность
ready = client.ready()
print("✅ ready:", ready)
assert ready["status"] == "ok", "Модель не готова!"

# 4. Делаем прогноз для тестовой смены
result = client.predict(shift_id="s_1", pool_size=10)
print("✅ predict:", result)

# Проверка: результат — список из ≤10 user_id
assert isinstance(result, list), "Ожидался список"
assert len(result) <= 10, "Слишком много кандидатов"
print("🎉 Всё работает!")

client.close()
```
