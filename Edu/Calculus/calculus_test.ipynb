$f(x_1, x_2, x_3) = \frac{1}{1 + e^{-2(x_1 + x_2 + x_3)}}$

```python
from sympy import Symbol, S, symbols
from sympy.calculus.util import function_range
x, x1, x2, x3 = symbols('x x1 x2 x3')
x = x1 + x2 + x3
f = 1 / (1 + exp(-2 * (x)) )
f1 = 1 / (1 + exp(-2 * (x1)) )
function_range(f, x, S.Reals)
```

# Производная / Derivative

Вычислить производную для функции $f(x) = x^4 +3x^3+4x^2+8$

```python
import sympy
x = sympy.Symbol('x')
expr = x**4 + 3*x**3 + 4*x**2 + 8
expr.diff(x)
```

Вычислить производную для функции $f(x) = ln(x^2 + 5x) + e^{sin(x)}$

```python
import sympy
from sympy import diff, sin, exp, log, Symbol
x = Symbol('x')
expr = log(x**2 + 5*x) + exp(sin(x))
expr.diff()
```

С помощью библиотеки SymPy найдите производную для функции
$y = sin(3x + log(x))\cdot e^x$

```python
from sympy import Symbol, log, sin, exp, diff
x = Symbol('x')
expr = sin(3*x + log(x)) * exp(x)
expr.diff(x)
```
