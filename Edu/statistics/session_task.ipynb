Задание 2.4

В нормально распределенной совокупности 20% значений x меньше 10 и 40% значений x больше 15. Найдите стандартное отклонение данного распределения. Ответ запишите с точностью до двух знаков после точки (даже если это нули), в качестве разделителя используйте точку.

```python
import scipy.stats as stats

# 20% значений меньше 10 => P(X < 10) = 0.20
# 40% значений больше 15 => P(X > 15) = 0.40 => P(X < 15) = 0.60

# Находим z-значения для стандартного нормального распределения
z1 = stats.norm.ppf(0.20)  # для 20%
z2 = stats.norm.ppf(0.60)  # для 60% (т.к. P(X < 15) = 0.60)

# Формируем систему уравнений:
# (10 - μ) / σ = z1  => 10 - μ = z1 * σ
# (15 - μ) / σ = z2  => 15 - μ = z2 * σ

# Вычитаем первое уравнение из второго:
# (15 - μ) - (10 - μ) = z2 * σ - z1 * σ
# 5 = (z2 - z1) * σ

sigma = 5 / (z2 - z1)

# Округляем до двух знаков после запятой
sigma_rounded = round(sigma, 2)

# Форматируем вывод
result = format(sigma_rounded, '.2f')
print(f"Стандартное отклонение σ = {result}")
```

Ответ запишите с точностью до двух знаков после точки (даже если это нули), в качестве разделителя используйте точку.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# --- Входные данные ---
n = 10  # Общее количество бросков
k = 7   # Требуемое количество попаданий
p = 0.7 # Вероятность попадания при одном броске (70%)

# --- Расчет вероятности ---

# Используем функцию binom.pmf (probability mass function) из scipy.stats
probability = binom.pmf(k, n, p)

# Форматируем ответ до двух знаков после запятой с точкой в качестве разделителя
formatted_answer = f"{probability:.2f}"

print(f"Вероятность того, что игрок попадет ровно {k} раз из {n} бросков: {formatted_answer}")

# --- Визуализация биномиального распределения ---

# Генерируем все возможные количества попаданий (от 0 до 10)
x = np.arange(0, n + 1)

# Рассчитываем вероятности для каждого из этих исходов
probabilities = binom.pmf(x, n, p)

# Строим график
plt.figure(figsize=(10, 6))
plt.bar(x, probabilities, color='skyblue', edgecolor='black')

# Выделяем столбец, соответствующий нашему ответу (k=7)
plt.bar(k, binom.pmf(k, n, p), color='tomato', edgecolor='black', label=f'P(X={k}) = {formatted_answer}')

plt.title('Биномиальное распределение вероятностей попаданий (n=10, p=0.7)')
plt.xlabel('Количество попаданий в корзину (k)')
plt.ylabel('Вероятность P(X=k)')
plt.xticks(x)
plt.legend()
plt.grid(axis='y', linestyle='--')
plt.show()
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# Параметры задачи
n = 10  # количество бросков
k = 7   # количество попаданий
p = 0.7 # вероятность попадания при одном броске

# Функция для вычисления биномиального коэффициента
def comb(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Вычисляем вероятность P(X = 7)
probability = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

# Округляем до двух знаков после запятой
probability_rounded = round(probability, 2)

# Форматируем результат
result = format(probability_rounded, '.2f')
print(f"Вероятность попасть 7 раз из 10 бросков: {result}")

# Дополнительно: Построение графика биномиального распределения
x = list(range(0, n+1))  # все возможные значения k (0 до 10)
probabilities = [comb(n, i) * (p ** i) * ((1 - p) ** (n - i)) for i in x]

# Создаем график
plt.figure(figsize=(12, 8))

# Столбчатая диаграмма для вероятностей
bars = plt.bar(x, probabilities, color='skyblue', edgecolor='black', alpha=0.7, label=f'Биномиальное распределение (n={n}, p={p})')

# Подсвечиваем столбец для k=7 другим цветом
for i, bar in enumerate(bars):
    if i == k:
        bar.set_color('red')
        bar.set_alpha(0.9)

# Добавляем значения вероятностей над столбцами
for i, prob in enumerate(probabilities):
    plt.text(i, prob + 0.005, f'{prob:.3f}', ha='center', va='bottom', fontsize=9)

# Настраиваем график
plt.xlabel('Количество попаданий (k)', fontsize=12)
plt.ylabel('Вероятность P(X = k)', fontsize=12)
plt.title(f'Биномиальное распределение для баскетболиста (n={n}, p={p})', fontsize=14, pad=20)
plt.xticks(x)
plt.grid(axis='y', alpha=0.3, linestyle='--')

# Добавляем легенду
plt.legend(fontsize=10)

# Показываем график
plt.tight_layout()
plt.show()

# Дополнительная информация
print(f"\nДополнительная информация:")
print(f"Среднее количество попаданий: μ = n * p = {n} * {p} = {n * p:.1f}")
print(f"Стандартное отклонение: σ = √(n * p * (1-p)) = √({n} * {p} * {1-p}) = {math.sqrt(n * p * (1 - p)):.2f}")
print(f"Наиболее вероятное количество попаданий: {int((n + 1) * p)}")

# Вероятности для всех возможных k
print(f"\nВероятности для всех k:")
for i, prob in enumerate(probabilities):
    print(f"  P(X = {i}) = {prob:.4f}")
```

Задание 2.6

Диаметр шара х измерен приближенно, причем 1.7 см ≤ x ≤ 1.8 см. Рассматривая диаметр шара как случайную величину, распределенную равномерно в интервале (a, b), найти математическое ожидание объема шара.
```python
import math
from sympy import symbols, integrate, pi, N, solveset, Interval

# --- 1. Определение параметров задачи ---

a = 1.7  # Нижняя граница интервала диаметра (см)
b = 1.8  # Верхняя граница интервала диаметра (см)

# --- 2. Символьное определение переменных и функций ---

# Определяем x как символьную переменную для диаметра
x = symbols('x')

# Определяем функцию объема V(x)
# V(x) = (1/6) * pi * x^3
V_x = (pi / 6) * x**3

# Определяем плотность вероятности f(x) для равномерного распределения
# f(x) = 1 / (b - a)
f_x_val = 1 / (b - a)
f_x = f_x_val

# --- 3. Вычисление математического ожидания (E[V]) ---

# Математическое ожидание вычисляется как интеграл от V(x) * f(x) в интервале [a, b]
# E[V] = integral from a to b of (V(x) * f(x)) dx
integrand = V_x * f_x

# Используем SymPy для символьного вычисления интеграла
# integrate(выражение, (переменная, нижний_предел, верхний_предел))
E_V_symbolic = integrate(integrand, (x, a, b))

print(f"Плотность вероятности f(x) = {f_x_val}\n")
print(f"Функция объема V(x) = {V_x}\n")
print(f"Подынтегральное выражение = {integrand}\n")
print(f"Символьное (точное) значение матожидания E[V] = {E_V_symbolic}\n")

# --- 4. Численное вычисление и округление результата ---

# N() используется для получения численного значения из символьного выражения
# Округляем результат до двух знаков после запятой, как того требует задача
E_V_numeric = N(E_V_symbolic)
result = round(E_V_numeric, 2)

print("-" * 30)
print(f"Численное значение E[V] (с плавающей точкой): {E_V_numeric}")
print(f"Округленный результат (с точностью до двух знаков): {result}")
print("-" * 30)

# --- 5. Вывод ответа в требуемом формате ---
# Формат требует точку в качестве разделителя и две цифры после точки.

# Используем форматирование строк Python для обеспечения нужного вывода
answer_formatted = f"{result:.2f}"

print(f"\nОтвет в требуемом формате: {answer_formatted}")
```

```python
import numpy as np
```
```python
import scipy.stats as stats
import matplotlib.pyplot as plt

# --- Исходные данные ---
n = 10                  # Объем выборки
mean_sample = 50        # Выборочное среднее
var_sample = 30         # Выборочная дисперсия
confidence_level = 0.95 # Уровень надежности (доверительная вероятность)

# --- Расчет параметров ---
df = n - 1  # Степени свободы

# Исправленное (несмещенное) среднее квадратическое отклонение (s)
# В scipy.stats.t.interval параметр scale ожидает стандартную ошибку среднего (SEM),
# которая уже учитывает поправку на несмещенность, если использовать np.std(ddof=1)
# или stats.sem().
# Мы рассчитаем вручную, как в предыдущем ответе:
s_sq = (n / (n - 1)) * var_sample
s = np.sqrt(s_sq)
sem = s / np.sqrt(n) # Стандартная ошибка среднего

# Расчет доверительного интервала с использованием функции scipy.stats.t.interval
# Функция interval возвращает нижнюю и верхнюю границы
lower_bound, upper_bound = stats.t.interval(
    confidence=confidence_level,
    df=df,
    loc=mean_sample,
    scale=sem
)

print(f"Нижняя граница (ручной расчет): {lower_bound:.2f}")
print(f"Верхняя граница (ручной расчет): {upper_bound:.2f}")
print(f"Верхняя граница (в формате ответа): {upper_bound:.2f}")

# --- Визуализация t-распределения и интервала ---

# 1. Генерация точек для графика t-распределения
# Создаем диапазон значений вокруг среднего для построения кривой
x = np.linspace(mean_sample - 4 * s, mean_sample + 4 * s, 1000)
# Рассчитываем плотность вероятности (PDF) для этих значений
pdf = stats.t.pdf(x, df, loc=mean_sample, scale=sem)

# 2. Построение графика
plt.figure(figsize=(10, 6))
plt.plot(x, pdf, 'k-', lw=2, label=f't-распределение (df={df})')

# 3. Заливка области доверительного интервала
# Используем plt.fill_between для выделения нужного диапазона
plt.fill_between(x, 0, pdf, where=(x >= lower_bound) & (x <= upper_bound),
                 color='skyblue', alpha=0.6, label=f'{confidence_level*100:.0f}% Доверительный интервал')

# 4. Добавление меток и заголовка
plt.axvline(mean_sample, color='red', linestyle='--', label=f'Выборочное среднее ({mean_sample})')
plt.axvline(lower_bound, color='gray', linestyle=':', label=f'Нижняя/Верхняя границы ({lower_bound:.2f}, {upper_bound:.2f})')
plt.axvline(upper_bound, color='gray', linestyle=':')
plt.title(f'Доверительный интервал для истинного среднего (t-распределение)')
plt.xlabel('Значение измеряемой величины')
plt.ylabel('Плотность вероятности (PDF)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import numpy as np
from scipy.stats import chi2

# Исходные данные
s = 0.8  # выборочное среднее квадратическое отклонение
n = 10   # объем выборки
confidence_level = 0.95 # уровень надежности

# Расчет параметров
df = n - 1  # степени свободы (degrees of freedom)
alpha = 1 - confidence_level # уровень значимости
# Для двустороннего интервала нужны квантили alpha/2 и 1-alpha/2

# Нахождение критических значений хи-квадрат (квантелей)
# Обратите внимание, что для стандартного отклонения используется "обратная" логика 
# при подстановке квантилей в формулу.
# Нижняя граница интервала использует верхний квантиль (1 - alpha/2), 
# а верхняя граница использует нижний квантиль (alpha/2).

chi2_lower_crit = chi2.ppf(alpha / 2, df)
chi2_upper_crit = chi2.ppf(1 - alpha / 2, df) # Это значение 19.023, как в ручном расчете

# Расчет границ доверительного интервала для генерального стандартного отклонения (sigma)
# Формула: sqrt[((n-1)*s^2) / chi2_crit]

lower_bound = np.sqrt(((n - 1) * s**2) / chi2_upper_crit)
upper_bound = np.sqrt(((n - 1) * s**2) / chi2_lower_crit)

# Форматирование вывода в требуемый формат: "нижняя_граница;верхняя_граница" с точностью до двух знаков
result_string = f"{lower_bound:.2f};{upper_bound:.2f}"

print(result_string)

# Для проверки также можно вывести границы с полной точностью:
# print(f"Нижняя граница: {lower_bound}")
# print(f"Верхняя граница: {upper_bound}")
```

```python
import scipy.stats as stats
import numpy as np
import matplotlib.pyplot as plt

# Исходные данные
regular = np.array([4.17, 5.58, 5.18, 6.11, 4.50, 4.61, 5.17, 4.53, 5.33, 5.14])
partial = np.array([4.52, 4.18, 4.84, 5.89, 3.57, 3.91, 5.96, 6.09, 4.74, 5.26])
none = np.array([4.81, 4.17, 4.41, 3.59, 5.87, 3.83, 6.03, 4.89, 4.32, 4.69])

# Проведение однофакторного ANOVA
f_stat, p_value = stats.f_oneway(regular, partial, none)

print(f"F-статистика: {f_stat:.4f}")
print(f"p-value: {p_value:.4f}")
print(f"Ответ: {p_value:.4f}")

# Проверка предположений ANOVA (необязательно для ответа, но полезно)
# Проверка на нормальность (тест Шапиро-Уилка) - для каждой группы
print("\nПроверка нормальности распределения (тест Шапиро-Уилка):")
for name, data in [("Регулярный уход", regular), 
                   ("Частичный уход", partial), 
                   ("Без ухода", none)]:
    shapiro_stat, shapiro_p = stats.shapiro(data)
    print(f"{name}: статистика={shapiro_stat:.4f}, p={shapiro_p:.4f}")

# Проверка равенства дисперсий (тест Левена)
levene_stat, levene_p = stats.levene(regular, partial, none)
print(f"\nТест Левена на равенство дисперсий: статистика={levene_stat:.4f}, p={levene_p:.4f}")

# Визуализация данных
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# График 1: диаграммы размаха
box_data = [regular, partial, none]
box_labels = ['Регулярный', 'Частичный', 'Без ухода']
axes[0].boxplot(box_data, tick_labels=box_labels)
axes[0].set_ylabel('Урожайность')
axes[0].set_title('Сравнение урожайности при разных типах ухода')
axes[0].grid(True, alpha=0.3)

# Добавляем средние значения
means = [np.mean(regular), np.mean(partial), np.mean(none)]
for i, mean in enumerate(means):
    axes[0].plot(i+1, mean, 'ro', markersize=8)
    axes[0].text(i+1.1, mean, f'{mean:.2f}', va='center')

# График 2: гистограммы распределений
colors = ['blue', 'orange', 'green']
for i, (data, label, color) in enumerate(zip(box_data, box_labels, colors)):
    axes[1].hist(data, alpha=0.6, label=label, color=color, bins=7, edgecolor='black')
axes[1].set_xlabel('Урожайность')
axes[1].set_ylabel('Частота')
axes[1].set_title('Гистограммы распределений')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# График 3: распределение F-статистики и критическая область
df_between = 2  # между группами: k-1, где k=3
df_within = len(regular) + len(partial) + len(none) - 3  # внутри групп: N-k
x = np.linspace(0, 8, 1000)
y = stats.f.pdf(x, df_between, df_within)

axes[2].plot(x, y, 'b-', linewidth=2, label=f'F({df_between},{df_within})')
axes[2].axvline(x=f_stat, color='red', linestyle='--', linewidth=2, label=f'F={f_stat:.2f}')

# Критическое значение для alpha=0.05
f_crit = stats.f.ppf(0.95, df_between, df_within)
x_fill = np.linspace(f_crit, 8, 200)
y_fill = stats.f.pdf(x_fill, df_between, df_within)
axes[2].fill_between(x_fill, y_fill, color='red', alpha=0.3, label=f'Критическая область\n(alpha=0.05)')

axes[2].set_xlabel('F-статистика')
axes[2].set_ylabel('Плотность вероятности')
axes[2].set_title(f'Распределение F и p-value={p_value:.4f}')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Дополнительная информация
print(f"\nСредние значения:")
print(f"Регулярный уход: {np.mean(regular):.4f}")
print(f"Частичный уход: {np.mean(partial):.4f}")
print(f"Без ухода: {np.mean(none):.4f}")
print(f"\nСтандартные отклонения:")
print(f"Регулярный уход: {np.std(regular, ddof=1):.4f}")
print(f"Частичный уход: {np.std(partial, ddof=1):.4f}")
print(f"Без ухода: {np.std(none, ddof=1):.4f}")
print(f"\nРазмеры выборок: n={len(regular)} для каждой группы")
print(f"Критическое значение F (alpha=0.05): {f_crit:.4f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

# Данные из таблицы
data = {
    'S': [2310, 2333, 2356, 2379, 2402, 2425, 2448, 2471, 2494, 2517, 2540, 2543, 2602, 2615, 2625],
    'O': [2, 2, 3, 3, 2, 4, 2, 2, 3, 4, 2, 3, 4, 2, 4],
    'E': [2, 2, 2, 2, 3, 3, 2, 2, 3, 4, 3, 4, 2, 3, 4],
    'Y': [20, 12, 33, 43, 53, 23, 37, 34, 23, 55, 22, 41, 29, 17, 20],
    'C': [142000, 144000, 151000, 151000, 139000, 169000, 126000, 142000, 163000, 169000, 149000, 152000, 156000, 162000, 172000]
}

# Создаем DataFrame
df = pd.DataFrame(data)
X = df[['S', 'O', 'E', 'Y']]
y = df['C']

# Способ 1: Ручная стандартизация (как требуется в задании)
# Стандартизация через вычитание среднего и деление на стандартное отклонение
X_standardized_manual = (X - X.mean()) / X.std()
y_standardized_manual = (y - y.mean()) / y.std()

# Строим модель линейной регрессии для стандартизованных переменных
model_manual = LinearRegression(fit_intercept=True)
model_manual.fit(X_standardized_manual, y_standardized_manual)

# Получаем коэффициенты регрессии
coefficients_manual = model_manual.coef_
intercept_manual = model_manual.intercept_

print("=" * 60)
print("МОДЕЛЬ С РУЧНОЙ СТАНДАРТИЗАЦИЕЙ")
print("=" * 60)
print("Средние значения переменных:")
print(f"  Площадь (S): {X['S'].mean():.2f}")
print(f"  Офисы (O): {X['O'].mean():.2f}")
print(f"  Входы (E): {X['E'].mean():.2f}")
print(f"  Эксплуатация (Y): {X['Y'].mean():.2f}")
print(f"  Цена (C): {y.mean():.2f}")
print("\nСтандартные отклонения:")
print(f"  Площадь (S): {X['S'].std():.2f}")
print(f"  Офисы (O): {X['O'].std():.2f}")
print(f"  Входы (E): {X['E'].std():.2f}")
print(f"  Эксплуатация (Y): {X['Y'].std():.2f}")
print(f"  Цена (C): {y.std():.2f}")

print("\nСтандартизованные коэффициенты регрессии:")
print(f"  Площадь (S): {coefficients_manual[0]:.4f}")
print(f"  Офисы (O): {coefficients_manual[1]:.4f}")
print(f"  Входы (E): {coefficients_manual[2]:.4f}")
print(f"  Эксплуатация (Y): {coefficients_manual[3]:.4f}")
print(f"  Intercept: {intercept_manual:.4f}")

# Способ 2: Использование StandardScaler (для проверки)
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_standardized_scaler = scaler_X.fit_transform(X)
y_standardized_scaler = scaler_y.fit_transform(y.values.reshape(-1, 1)).flatten()

model_scaler = LinearRegression(fit_intercept=True)
model_scaler.fit(X_standardized_scaler, y_standardized_scaler)

print("\n" + "=" * 60)
print("МОДЕЛЬ СО STANDARDSCALER (ДЛЯ ПРОВЕРКИ)")
print("=" * 60)
print("Стандартизованные коэффициенты регрессии:")
print(f"  Площадь (S): {model_scaler.coef_[0]:.4f}")
print(f"  Офисы (O): {model_scaler.coef_[1]:.4f}")
print(f"  Входы (E): {model_scaler.coef_[2]:.4f}")
print(f"  Эксплуатация (Y): {model_scaler.coef_[3]:.4f}")
print(f"  Intercept: {model_scaler.intercept_:.4f}")

# Сравнение коэффициентов по модулю
abs_coefficients = np.abs(coefficients_manual)
max_abs_coefficient = np.max(abs_coefficients)
max_abs_index = np.argmax(abs_coefficients)

variable_names = ['Площадь (S)', 'Офисы (O)', 'Входы (E)', 'Эксплуатация (Y)']
max_variable = variable_names[max_abs_index]

print("\n" + "=" * 60)
print("АНАЛИЗ КОЭФФИЦИЕНТОВ")
print("=" * 60)
for i, name in enumerate(variable_names):
    print(f"{name}: |{coefficients_manual[i]:.4f}| = {abs_coefficients[i]:.4f}")

print(f"\nМаксимальный по модулю коэффициент: {max_abs_coefficient:.4f}")
print(f"Соответствует переменной: {max_variable}")
print(f"Знак коэффициента: {'положительный' if coefficients_manual[max_abs_index] > 0 else 'отрицательный'}")

# Интерпретация
print("\n" + "=" * 60)
print("ИНТЕРПРЕТАЦИЯ СТАНДАРТИЗОВАННЫХ КОЭФФИЦИЕНТОВ")
print("=" * 60)
print("Стандартизованные коэффициенты показывают, на сколько стандартных отклонений")
print("изменится цена (в стандартных отклонениях) при изменении предиктора")
print("на одно стандартное отклонение, при условии, что другие переменные постоянны.")

for i, name in enumerate(variable_names):
    coeff = coefficients_manual[i]
    interpretation = "увеличивается" if coeff > 0 else "уменьшается"
    print(f"\n{name}: {coeff:.4f}")
    print(f"  При увеличении {name.lower()} на 1 стандартное отклонение,")
    print(f"  цена в среднем {interpretation} на {abs(coeff):.4f} стандартных отклонений.")

print("\n" + "=" * 60)
print(f"ОТВЕТ: Максимальный (по модулю) коэффициент = {max_abs_coefficient:.4f}")
print("=" * 60)

# Дополнительно: сравнение с нестандартизованной моделью
print("\n" + "=" * 60)
print("СРАВНЕНИЕ С НЕСТАНДАРТИЗОВАННОЙ МОДЕЛЬЮ")
print("=" * 60)
model_raw = LinearRegression(fit_intercept=True)
model_raw.fit(X, y)

print("Нестандартизованные коэффициенты:")
print(f"  Площадь (S): {model_raw.coef_[0]:.4f}")
print(f"  Офисы (O): {model_raw.coef_[1]:.4f}")
print(f"  Входы (E): {model_raw.coef_[2]:.4f}")
print(f"  Эксплуатация (Y): {model_raw.coef_[3]:.4f}")
print(f"  Intercept: {model_raw.intercept_:.4f}")

# Проверка: преобразование нестандартизованных коэффициентов в стандартизованные
# beta_standardized = beta_raw * (std_X / std_y)
std_X = X.std()
std_y = y.std()
beta_converted = model_raw.coef_ * (std_X / std_y)
print("\nПреобразованные нестандартизованные коэффициенты:")
print(f"  Площадь (S): {beta_converted[0]:.4f}")
print(f"  Офисы (O): {beta_converted[1]:.4f}")
print(f"  Входы (E): {beta_converted[2]:.4f}")
print(f"  Эксплуатация (Y): {beta_converted[3]:.4f}")
print("\nСовпадают со стандартизованными: ", np.allclose(beta_converted, coefficients_manual, rtol=1e-4))
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Данные из таблицы
data = {
    'S': [2310, 2333, 2356, 2379, 2402, 2425, 2448, 2471, 2494, 2517, 
          2540, 2543, 2602, 2615, 2625],
    'O': [2, 2, 3, 3, 2, 4, 2, 2, 3, 4, 2, 3, 4, 2, 4],
    'E': [2, 2, 2, 2, 3, 3, 2, 2, 3, 4, 3, 4, 2, 3, 4],
    'Y': [20, 12, 33, 43, 53, 23, 37, 34, 23, 55, 22, 41, 29, 17, 20],
    'C': [142000, 144000, 151000, 151000, 139000, 169000, 126000, 142000, 
          163000, 169000, 149000, 152000, 156000, 162000, 172000]
}

# Создаем DataFrame
df = pd.DataFrame(data)

# Определяем независимые и зависимые переменные
X = df[['S', 'O', 'E', 'Y']]
y = df['C']

# Добавляем константу (intercept) для модели
X_with_const = sm.add_constant(X)

# Строим модель линейной регрессии
model = sm.OLS(y, X_with_const).fit()

# Получаем коэффициенты регрессии
coefficients = model.params

print("=" * 60)
print("ЛИНЕЙНАЯ РЕГРЕССИОННАЯ МОДЕЛЬ (БЕЗ СТАНДАРТИЗАЦИИ)")
print("=" * 60)
print("Коэффициенты регрессии:")
print(f"  Intercept: {coefficients['const']:.4f}")
print(f"  Площадь (S): {coefficients['S']:.4f}")
print(f"  Офисы (O): {coefficients['O']:.4f}")
print(f"  Входы (E): {coefficients['E']:.4f}")
print(f"  Эксплуатация (Y): {coefficients['Y']:.4f}")

# Дополнительная информация о модели
print("\n" + "=" * 60)
print("КАЧЕСТВО МОДЕЛИ")
print("=" * 60)
print(f"R-квадрат (R²): {model.rsquared:.4f}")
print(f"Скорректированный R²: {model.rsquared_adj:.4f}")
print(f"F-статистика: {model.fvalue:.4f}")
print(f"p-value F-теста: {model.f_pvalue:.4f}")

# Выводим детальный отчет
print("\n" + "=" * 60)
print("ДЕТАЛЬНЫЙ ОТЧЕТ ПО МОДЕЛИ")
print("=" * 60)
print(model.summary())

# Проверка с использованием numpy (для верификации)
print("\n" + "=" * 60)
print("ПРОВЕРКА ЧЕРЕЗ NUMPY")
print("=" * 60)
# Используем метод наименьших квадратов
X_np = np.column_stack([np.ones(len(X)), X])
beta = np.linalg.lstsq(X_np, y, rcond=None)[0]
print(f"Коэффициенты (numpy):")
print(f"  Intercept: {beta[0]:.4f}")
print(f"  Площадь (S): {beta[1]:.4f}")
print(f"  Офисы (O): {beta[2]:.4f}")
print(f"  Входы (E): {beta[3]:.4f}")
print(f"  Эксплуатация (Y): {beta[4]:.4f}")

print("\n" + "=" * 60)
print("ИНТЕРПРЕТАЦИЯ КОЭФФИЦИЕНТА ПРИ ПЛОЩАДИ")
print("=" * 60)
print(f"Коэффициент при площади: {coefficients['S']:.4f}")
print("Это означает, что при увеличении площади на 1 м²,")
print(f"цена здания в среднем увеличивается на {coefficients['S']:.2f} у.е.,")
print("при условии, что другие факторы (офисы, входы, возраст) остаются постоянными.")

print("\n" + "=" * 60)
print(f"ОТВЕТ: Коэффициент при площади = {coefficients['S']:.4f}")
print("=" * 60)

# Для наглядности выведем уравнение модели
print("\nУРАВНЕНИЕ РЕГРЕССИИ:")
print(f"Цена = {coefficients['const']:.2f} + {coefficients['S']:.2f}×S + " +
      f"{coefficients['O']:.2f}×O + {coefficients['E']:.2f}×E + " +
      f"{coefficients['Y']:.2f}×Y")
```

Задание 5.4

Оцените значимость уравнения в целом. В ответе укажите разность расчетного F-критерия уравнения регрессии и табличного значения при заданных параметрах, то есть: Fвыч – Fтабл. Получение отрицательного числа свидетельствует, что уравнение незначимо. Ответ запишите с точностью до двух знаков после точки (даже если это нули), в качестве разделителя используйте точку.

```python
import pandas as pd
import numpy as np
import scipy.stats as stats

# Данные
data = {
    'S': [2310, 2333, 2356, 2379, 2402, 2425, 2448, 2471, 2494, 2517, 
          2540, 2543, 2602, 2615, 2625],
    'O': [2, 2, 3, 3, 2, 4, 2, 2, 3, 4, 2, 3, 4, 2, 4],
    'E': [2, 2, 2, 2, 3, 3, 2, 2, 3, 4, 3, 4, 2, 3, 4],
    'Y': [20, 12, 33, 43, 53, 23, 37, 34, 23, 55, 22, 41, 29, 17, 20],
    'C': [142000, 144000, 151000, 151000, 139000, 169000, 126000, 142000, 
          163000, 169000, 149000, 152000, 156000, 162000, 172000]
}

df = pd.DataFrame(data)
X = df[['S', 'O', 'E', 'Y']]
y = df['C']

# Построение модели для получения R²
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)

# Расчет F-критерия
n = len(y)  # 15 наблюдений
k = X.shape[1] + 1  # 4 признака + константа = 5 параметров
r_squared = model.score(X, y)

# F-критерий расчетный
F_calc = (r_squared / (k - 1)) / ((1 - r_squared) / (n - k))

# Табличное значение F-критерия
alpha = 0.05  # уровень значимости 5%
df1 = k - 1   # степени свободы числителя = 4
df2 = n - k   # степени свободы знаменателя = 10

F_tab = stats.f.ppf(1 - alpha, df1, df2)

# Разность
difference = F_calc - F_tab

print("=" * 60)
print("ОЦЕНКА ЗНАЧИМОСТИ УРАВНЕНИЯ РЕГРЕССИИ В ЦЕЛОМ")
print("=" * 60)
print(f"Количество наблюдений (n): {n}")
print(f"Количество параметров (k): {k}")
print(f"Коэффициент детерминации (R²): {r_squared:.4f}")
print()
print(f"Расчетный F-критерий (Fвыч): {F_calc:.4f}")
print(f"Табличное значение (Fтабл) при α={alpha}: {F_tab:.4f}")
print(f"Степени свободы: df1={df1}, df2={df2}")
print()
print(f"Разность: Fвыч - Fтабл = {F_calc:.4f} - {F_tab:.4f} = {difference:.4f}")
print(f"Округление до двух знаков после точки: {difference:.2f}")

# Проверка значимости
if F_calc > F_tab:
    print("\nВывод: Уравнение регрессии статистически значимо в целом")
    print(f"(Fвыч = {F_calc:.4f} > Fтабл = {F_tab:.4f})")
else:
    print("\nВывод: Уравнение регрессии незначимо в целом")
    print(f"(Fвыч = {F_calc:.4f} ≤ Fтабл = {F_tab:.4f})")

print("=" * 60)

# Дополнительно: расчет с использованием statsmodels для проверки
import statsmodels.api as sm
X_const = sm.add_constant(X)
model_sm = sm.OLS(y, X_const).fit()
print("\nПроверка через statsmodels:")
print(f"F-статистика: {model_sm.fvalue:.4f}")
print(f"p-value F-теста: {model_sm.f_pvalue:.6f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# Данные
data = {
    'S': [2310, 2333, 2356, 2379, 2402, 2425, 2448, 2471, 2494, 2517, 
          2540, 2543, 2602, 2615, 2625],
    'O': [2, 2, 3, 3, 2, 4, 2, 2, 3, 4, 2, 3, 4, 2, 4],
    'E': [2, 2, 2, 2, 3, 3, 2, 2, 3, 4, 3, 4, 2, 3, 4],
    'Y': [20, 12, 33, 43, 53, 23, 37, 34, 23, 55, 22, 41, 29, 17, 20],
    'C': [142000, 144000, 151000, 151000, 139000, 169000, 126000, 142000, 
          163000, 169000, 149000, 152000, 156000, 162000, 172000]
}

df = pd.DataFrame(data)
X = df[['S', 'O', 'E', 'Y']]
y = df['C']

# Построение модели
model = LinearRegression()
model.fit(X, y)

# Коэффициент детерминации R²
r_squared = model.score(X, y)

# Доля необъясненной дисперсии
unexplained_variance = 1 - r_squared

# В процентах
unexplained_percent = unexplained_variance * 100

print("=" * 50)
print("АНАЛИЗ ОБЪЯСНЕННОЙ ДИСПЕРСИИ")
print("=" * 50)
print(f"Коэффициент детерминации R²: {r_squared:.4f}")
print(f"Объясненная дисперсия: {r_squared * 100:.2f}%")
print(f"Необъясненная дисперсия (1 - R²): {unexplained_variance:.4f}")
print(f"Необъясненная дисперсия в процентах: {unexplained_percent:.2f}%")
print(f"Округление до одного знака после точки: {unexplained_percent:.1f}%")
print("=" * 50)

# Проверка через остатки
y_pred = model.predict(X)
ss_res = np.sum((y - y_pred) ** 2)  # сумма квадратов остатков (необъясненная вариация)
ss_tot = np.sum((y - np.mean(y)) ** 2)  # общая сумма квадратов

print("\nПроверка расчетом через суммы квадратов:")
print(f"SS_res (необъясненная вариация): {ss_res:.2f}")
print(f"SS_tot (общая вариация): {ss_tot:.2f}")
print(f"SS_res / SS_tot: {ss_res/ss_tot:.4f} = {(ss_res/ss_tot)*100:.2f}%")
```

Будем считать, что вероятность наступления события Х в n испытаниях подчиняется биномиальному распределению. Найти методом максимального правдоподобия оценку параметра p биномиального распределения, если в 16 независимых испытаниях событие Х появилось 10 раз и в 34 независимых испытаниях событие Х появилось 11 раз. Ответ запишите с точностью до двух знаков после точки (даже если это нули), в качестве разделителя используйте точку.

```python
# Данные из условия задачи
trials = [(16, 10), (34, 11)]  # (n, k)

def calculate_mle_p(data):
    total_n = sum(n for n, k in data)
    total_k = sum(k for n, k in data)
    
    # Оценка максимального правдоподобия p = k / n
    p_hat = total_k / total_n
    return p_hat

# Расчет
result = calculate_mle_p(trials)

# Вывод результата с точностью до двух знаков
print(f"{result:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
from scipy.optimize import minimize_scalar

# Данные
k1, n1 = 10, 16  # 10 успехов в 16 испытаниях
k2, n2 = 11, 34  # 11 успехов в 34 испытаниях

# Метод максимального правдоподобия для биномиального распределения
# Функция правдоподобия: L(p) = p^{k1+k2} * (1-p)^{n1+n2-k1-k2}
# Логарифмическое правдоподобие: l(p) = (k1+k2)*log(p) + (n1+n2-k1-k2)*log(1-p)
# Максимум достигается при p = (k1+k2)/(n1+n2)

# Оценка параметра p
p_est = (k1 + k2) / (n1 + n2)
print(f"Оценка параметра p: {p_est:.2f}")

# Построение графиков
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# График 1: Функция правдоподобия
p_values = np.linspace(0.01, 0.99, 1000)
likelihood = p_values**(k1+k2) * (1-p_values)**(n1+n2-k1-k2)

ax1.plot(p_values, likelihood, 'b-', linewidth=2, label='L(p)')
ax1.axvline(p_est, color='r', linestyle='--', 
            label=f'Оценка ММП: p = {p_est:.2f}')
ax1.set_xlabel('p (вероятность успеха)')
ax1.set_ylabel('Функция правдоподобия L(p)')
ax1.set_title('Функция правдоподобия для двух экспериментов')
ax1.grid(True, alpha=0.3)
ax1.legend()

# График 2: Логарифмическая функция правдоподобия
log_likelihood = (k1+k2)*np.log(p_values) + (n1+n2-k1-k2)*np.log(1-p_values)
ax2.plot(p_values, log_likelihood, 'g-', linewidth=2, 
         label='log L(p)')
ax2.axvline(p_est, color='r', linestyle='--', 
            label=f'Оценка ММП: p = {p_est:.2f}')
ax2.set_xlabel('p (вероятность успеха)')
ax2.set_ylabel('Логарифмическая функция правдоподобия')
ax2.set_title('Логарифмическая функция правдоподобия')
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()

# Дополнительная проверка численной оптимизацией
def neg_log_likelihood(p):
    # Отрицательное логарифмическое правдоподобие для минимизации
    if p <= 0 or p >= 1:
        return np.inf
    return -((k1+k2)*np.log(p) + (n1+n2-k1-k2)*np.log(1-p))

# Минимизируем отрицательное логарифмическое правдоподобие
result = minimize_scalar(neg_log_likelihood, bounds=(0.001, 0.999), 
                        method='bounded')
print(f"Численная проверка: p = {result.x:.5f}")
print(f"Совпадает с аналитическим решением: {np.abs(result.x - p_est) < 1e-10}")

# Вывод ответа
print(f"\nОтвет: {p_est:.2f}")
```

```python
import math

# Среднее количество пассажиров (Математическое ожидание E[X])
mean_passengers = 1700

# Пороговое значение для события (a)
threshold_a = 2101 # Используем 2101, так как нас интересует "не более 2100" (X <= 2100)

# Применение неравенства Маркова к противоположному событию P(X >= a)
# P(X >= a) <= E[X] / a
prob_ge_a_upper_bound = mean_passengers / threshold_a

# Минимальная вероятность искомого события P(X <= 2100) = 1 - P(X >= 2101)
# P(X <= a-1) >= 1 - (E[X] / a)
min_prob_le_a_minus_1 = 1 - prob_ge_a_upper_bound

# Форматирование результата до двух знаков после запятой
result_formatted = f"{min_prob_le_a_minus_1:.2f}"

print(f"Верхняя граница вероятности P(X >= {threshold_a}): {prob_ge_a_upper_bound:.4f}")
print(f"Минимальная вероятность P(X <= 2100): {min_prob_le_a_minus_1:.4f}")
print(f"Ответ в требуемом формате: {result_formatted}")
```
