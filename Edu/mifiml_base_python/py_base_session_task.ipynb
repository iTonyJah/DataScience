# Задание 1 (10 баллов)
Обратный порядок слов в блоках текста

## Дано:

Дан текстовый файл, каждое предложение которого занимает одну строку. Напишите программу, которая разделяет текст на блоки — каждый блок состоит из нескольких предложений. Затем переворачивает порядок слов только внутри каждого предложения, не меняя порядок самих предложений в блоке.

Файл содержит:

1. Привет как дела
2. На улице идет дождь
3. Я люблю программирование


Критерии:

- Корректное деление на блоки — 3 балла
- Переворот слов в предложениях — 5 баллов
- Структура вывода и читаемость — 2 балла

## Решение

```python
filename = 'py_base_session_task_test_reverse_words1.txt'

try:
    with open(filename, 'r', encoding='utf-8') as file:
        content = file.read()
except:
    content = """Привет как дела
На улице идет дождь
Я люблю программирование"""

original_row_list = content.split('\n')
reversed_row_list = []

for row in original_row_list:
    sentence = row.split()
    reverse_sentence = sentence[::-1]
    reverse_row = ' '.join(reverse_sentence)
    reversed_row_list.append(reverse_row)

reversed_content = '. '.join(reversed_row_list)

print(reversed_content)
```

# Задание 2 (6 баллов)
Сжатие строки

## Дано:

Напишите функцию, которая принимает строку и сжимает её определённым образом. Строки содержат произвольные символы, включая пробелы и спецсимволы, и требуют точного учета длины при кодировании. Сжатие строки происходит сериями одинаковых символов в формате символколичество, но только если длина сжатой строки не превышает исходную.

Примеры:

- вход: aaabbc → вывод: a3b2c,
- вход: abcd → вывод: abcd (так как сжатая равна).


Критерии:

- Правильная реализация логики сжатия — 3 балла
- Учет условия длины — 2 балла
- Обработка краевых случаев (пустые строки, одиночные символы) — 1 балл

## Решение

```python
def compress_string(string):
    compressed = []
    chars = []
    char_count = 0
    for char in string:
        if chars == [] or chars[-1] == char:
            chars.append(char)
            char_count += 1
            continue
        if char_count > 1:
            compressed.append(chars[-1] + str(char_count))
        else:
            compressed.append(''.join(chars))
        chars = [char]
        char_count = 1

    if char_count > 1:
        compressed.append(chars[-1] + str(char_count))
    else:
        compressed.append(''.join(chars))
    
    return ''.join(compressed)
    
```

Проверка скобок

Напишите функцию, которая проверяет корректность расстановки круглых, квадратных и фигурных скобок.

Критерии:
- Распознавание всех трёх типов скобок — 3 балла
- Учёт вложенности — 4 балла
- Чистота кода и читаемость — 2 балла
- Возврат булевого результата — 1 балл

```python
def check_brackets_close(code):
    
    open_brackets = {'(', '[', '{'}
    close_brackets = {')', ']', '}'}
    brackets = open_brackets | close_brackets

    #оставим только скобки в том же порядке
    bracket_string = [char for char in code if char in brackets]
    bracket_string = ''.join(bracket_string)

    #удалим парные скобки, в итоге останутся только некорректные
    while '()' in bracket_string or '[]' in bracket_string or '{}' in bracket_string:
        bracket_string = bracket_string.replace('()', '')
        bracket_string = bracket_string.replace('[]', '')
        bracket_string = bracket_string.replace('{}', '')        
    
    return False if bracket_string else True
        
```

# Задание 4 (10 баллов)
Генератор паролей

Напишите функцию, которая генерирует пароль заданной длины. В реализации надо учитывать, что:
1. Пароль должен содержать буквы, цифры и специальные символы.
2. Длина пароля задается пользователем.

Критерии:
- Корректная генерация длины — 3 балла
- Наличие всех типов символов — 4 балла
- Уникальность и случайность паролей — 2 балла
- Удобство использования функции — 1 балл

```python
def generate_password(password_length):
    import random

    def fill_minimum(symbols_list, length=4):
        if length == 4:
            password = random.choice(symbols_list[0])
            password += random.choice(symbols_list[1])
            password += random.choice(symbols_list[2])
            password += random.choice(symbols_list[3])
        elif length == 3:
            password = random.choice(symbols_list[0] + symbols_list[1])
            password += random.choice(symbols_list[2])
            password += random.choice(symbols_list[3])
        elif length == 2:
            password = random.choice(symbols_list[0] + symbols_list[1])
            password += random.choice(symbols_list[2] + symbols_list[3])
        elif length == 1:
            password = random.choice(symbols_list[0] + 
                                     symbols_list[1] +
                                     symbols_list[2] +
                                     symbols_list[3]
                                    )

        return password

    caps = 'QWERTYUIOPASDFGHJKLZXCVBNM'
    lows = caps.lower()
    digits = '1234567890'
    spec = '~!?@#$%^&*_-+()[]{}></|.,:;'
    symbols = caps + lows + digits + spec
    symbols_list = [caps, lows, digits, spec]

    password = fill_minimum(symbols_list, min(4, password_length))

    for i in range(password_length - 4):
        password += random.choice(symbols)

    password = list(password)
    random.shuffle(password)
    password = ''.join(password)
    
    return password
```

```python
password = generate_password(password_length=17)
print(password)
```

# Задание 5 (6 баллов)
Почти палиндром

Напишите программу, которая проверяет, является ли строка палиндромом или «почти палиндромом». «Почти палиндром» означает, что можно удалить одну букву, чтобы строка стала палиндромом.

Критерии:
- Проверка палиндромов — 3 балла
- Обработка случая «почти палиндром» — 2 балла
- Учет регистра и пробелов — 1 балл

```python
def is_palindrome(string):
    def check_palindrome(string):
        errors = 0
        for i, char in enumerate(string):
            if char != string[-i-1]:
                errors += 1
        return errors


    string = ''.join([char.lower() for char in string if char.isalpha()])

    if check_palindrome(string) == 0:
        return 'Строка палиндром.'
    elif check_palindrome(string) == 2:
        return 'Строка почти палиндром, достаточно исправить одну букву.'
    else:
        for i in range(len(string)):
            cut_string = string[:i] + string[i+1:]
            if check_palindrome(cut_string) == 0:
                break
        if check_palindrome(cut_string) == 0:
            return 'Строка почти палиндром, достаточно удалить одну букву.'
        else:
            return 'Строка не палиндром.'
    
```
