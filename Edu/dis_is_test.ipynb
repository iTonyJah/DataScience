```python
def test_int_str():
    x = 1000
    y = int('1000')
    return x is y

test_int_str()
```

```python
import dis
dis.dis(test_int_str)
```

```python
def test_int_lst():
    x = 1000
    y = list(range(1001))[-1]
    if x == y:
        return x is y

test_int_str()
```

```python
import dis
dis.dis(test_int_lst)
```

```python
def test_int_int():
    x = 1000
    y = 10**3
    return x is y

test_int_int()
```

```python
dis.dis(test_int_int)
```
