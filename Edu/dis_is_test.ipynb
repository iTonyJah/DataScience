```python
import dis
```

```python
def test():
    x = 1000
    y = 10**3 + 500 - 500
    return x is y

test(), test.__code__.co_consts, test.__code__.co_varnames
```

```python
dis.dis(test)
```

```python
def test():
    x = '1000'
    y = '10' + '0' * 2
    return x is y

test(), test.__code__.co_consts
```

```python
def test():
    x = [1000]
    y = [1000]
    return x is y

test(), test.__code__.co_consts, test.__code__.co_varnames
```

```python
dis.dis(test)
```

```python
def test():
    x = (1000, 1)
    y = (10**3, 2-1)
    return x is y

test(), test.__code__.co_consts
```

```python
def test():
    x = {1, 2, 3}
    y = {1, 2, 3}
    return x is y

test(), test.__code__.co_consts
```

```python
def test():
    x = 1000
    y = list(range(1000))[-1] + 1
    if x == y:
        return x is y

test(), test.__code__.co_consts
```

```python
dis.dis(test)
```
