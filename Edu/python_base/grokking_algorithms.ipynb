```python
def sum_digits_loop(number):
    total = 0
    for digit in str(number):
        total += int(digit)
    return total
```

```python
def sum_digits_recursive(number):
    if number == '':
        return 0
    return int(str(number)[:1]) + sum_digits_recursive(str(number)[1:])
```

```python
number = 1234567890**200
sum_digits_loop(number), sum_digits_recursive(number)
```
