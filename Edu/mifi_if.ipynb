Приоритет логических операторов ниже чем приоритет операторов сравнения

Таблица истинности оператора *not*:
|   x   | not x |
|:-----:|:-----:|
| True  | False |
| False | True  |

Таблица истинности оператора *and*:
| x     | y     | x and y |
|-------|-------|---------|
| False | False | False   |
| False | True  | False   |
| True  | False | False   |
| True  | True  | True    |

Таблица истинности оператора *or*:
| x     | y     | x or y |
|-------|-------|--------|
| False | False | False  |
| False | True  | True   |
| True  | False | True   |
| True  | True  | True   |

Операторы AND и OR являются сокращенными (short-circuit):

* OR возвращает TRUE, если первый аргумент равен TRUE, иначе возращает второй аргумент.

* AND возвращает FALSE, если первый аргумент равен FALSE, иначе возвращает второй аргумент.

```python
print(None or 0)
print(0 or 5)
print('python' or None)
print([0, 0] or [6, 9])
```

```python
print(None and 10)
print(5 and 0.0)
print('python' and {})
print([1, 2, 3] and [6, 9])
print([] and [6, 9])
```

```python
x = 0
y = 100

if x != 0:
    res3 = y // x
    print(f'{y} // {x} = {res3}')
else:
    print('На ноль делить нельзя!')
print('За пределами условного оператора')
```

```python
x = -10
if x > 0:
    print("x - положительное число")
    print('тогда умножим его на 5')
    x *= 5 # x = x * 5
elif not x: 
    # x == 0 --> not False = True
    print('x равен нулю')
else: # когда все условия не выполнились
    print("x - отрицательное число")
    x = abs(x)
print('конец проверки')
print(f'результат преобразования = {x}')
```

#### 'Моржовый' оператор

'Моржовый' оператор / оператор слияния (:=) позволяет присваивать значение переменной и проверять его в одном выражении. Обычно он используется в условиях оператора `if` или внутри выражений.

```python
if (n := len([1, 2, 3])) > 2:
    print(f"Список достаточно длинный ({n} элемента)")
```

```python
x = 17
y = 3
n = x % y
if not n: # n == 0
    print(f"{x} делится на {y} без остатка")
else:
    print(f"Остаток от деления {x} на {y} равен {n}")
```

```python
x = 17
y = 3

if (n := x % y) == 0:
    print(f"{x} делится на {y} без остатка")
else:
    print(f"Остаток от деления {x} на {y} равен {n}")
```

Моржовый оператор удобен, когда вам нужно выполнить присваивание значения и проверку условия одновременно, что может сделать ваш код более компактным и читаемым.

На вход поступает произвольное количество значений. Считывание заканчивается когда на вход поступает 0. Необходимо добавлять эти значения в список, если они являются целыми положительными числами.

```python
nums = []
num = input()

while num != '0':
    if num.isdigit():
        nums.append(int(num))
        print(f'Значение {num} добавлено в список.')
    num = input()
print(nums)
```

```python
nums = []

while (num := input()) != '0':
    if num.isdigit():
        nums.append(int(num))
        print(f'Значение {num} добавлено в список.')
print(nums)
```

Основная идея: match позволяет сравнивать значение (или даже структуру данных) с разными шаблонами (patterns) и выполнять соответствующий блок кода.

```python
number = 42
match number:
    case 42:
        print('answer')
    case 43:
        print('not answer')

if number == 42:
     print('answer')
elif number == 43:
    print('not answer')
```

```python
number = 0
match number:
    case 42:
        print("Its’s forty two")
    case 43:
        print('not answer')
    case _:
        print("I don’t know, what it is")
```

```python
# сопоставление с кортежами
point = (0, 5)

match point:
    case (0, y): # первый элемент кортежа равен 0, второй - любое значение
        print(f"Точка на оси Y, y={y}")
    case (x, 0):
        print(f"Точка на оси X, x={x}")
    case (x, y):
        print(f"Обычная точка: {x=}, {y=}")
```
