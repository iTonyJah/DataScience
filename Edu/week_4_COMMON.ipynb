## 🟢 1. PEP-8 и почему важно ему следовать.

**PEP-8 (Python Enhancement Proposal 8)** — это официальный стиль написания кода в Python.

**Почему важно следовать PEP-8?**
1. *Читаемость кода* — код легко понять коллегам и самому через время.
2. *Поддержка и масштабирование* — единый стиль облегчает совместную работу в команде.
3. *Инструменты автоматизации* — большинство IDE и линтеров умеют проверять PEP-8.
4. *Сообщество* — соблюдение стандарта облегчает участие в open-source проектах.

**Основные правила PEP-8**

1. *Отступы и пробелы*
    - Использовать **4 пробела** на уровень отступа.
    - Не использовать табы.
    - После запятой, двоеточия, точки с запятой ставить пробел.
2. *Длина строки*
    - Максимум **79 символов** на строку (для кода).
    - Максимум **72 символа** для комментариев и docstring.
3. *Именование*
    - Переменные и функции: `snake_case` (`my_variable`, `calculate_sum()`).
    - Классы: `CamelCase` (`MyClass`).
    - Константы: `UPPER_CASE` (`PI = 3.14`).
    - Модули и пакеты: короткие, lowercase, можно использовать `_` (`my_module`).
4. *Пробелы вокруг операторов*
    ```python
    x = 5             # правильно
    y=5               # неправильно
    ```
5. *Комментарии*
   - Комментарии должны быть осмысленными и помогать понять код.
   - Начинать комментарий с пробела после `#`.
   - Избегать излишних комментариев — лучше писать понятный код.
   - Для многострочных комментариев использовать блоки с `#` в начале каждой строки.

    ```python
    # Это правильный комментарий
    x = x + 1  # Увеличиваем счётчик на единицу
    ```
6. *Докстринги*
    * Используются для функций, классов и модулей.
    * Оформляются тройными кавычками `"""`.
    * В начале краткое описание идеи функции, далее аргументы и возвращаемое значение.
    * Если текст многострочный, то после "идеи" — пустая строка.

    ```python
    def add(a, b):
        """
        Складывает два числа.

        Args:
           a (int): Первое число.
           b (int): Второе число.

        Returns:
           int: Сумма чисел.
        """
        return a + b
```
7. *Импорты*
    * Каждый импорт — с новой строки.
    * Сначала стандартные библиотеки, потом сторонние, потом собственные модули.
    * Использовать абсолютные импорты, а не относительные.
    * Не использовать `import *`.

    ```python
    import os
    import sys

    import numpy as np
    import pandas as pd

    from my_package import my_module
    ```
8. *Пустые строки*
    * Между функциями и классами — **2 пустые строки**.
    * Внутри класса между методами — **1 пустая строка**.
    * Избегать лишних пустых строк.
9. *Выравнивание и перенос строк*
    * Аргументы функции при длинных вызовах переносятся с выравниванием.
    * Использовать висячую запятую (trailing comma) в многострочных структурах данных.
    ```python
    my_list = [
        1,
        2,
        3,
    ]

    result = my_function(
        arg1, arg2,
        kwarg1=10, kwarg2=20,
    )
    ```
10. *Прочие рекомендации*
    * Всегда использовать `if __name__ == "__main__":` для запуска кода.
    * Логические значения: `True`, `False`, `None` — обязательно с заглавной буквы.
    * Не сравнивать напрямую с `True/False/None`, вместо этого:
    ```python
    if my_var is None:      # правильно
    if my_var == None:      # неправильно
    ```
    
**PS: "8" в названии - это номер в реестре PEP, а не количество правил;)**

## 🟢 2. Модули и библиотеки.

**Модуль** - это отдельный файл с кодом на Python (`.py`), который можно подключить в другой программе.
- Обычно содержит функции, классы или переменные для повторного использования.
- Пример: `math`, `random`.

**Библиотека** - это набор модулей, объединённых общей задачей.
* Удобна для расширения возможностей Python (например, работа с данными, графикой, сетью).
* Примеры: `NumPy`, `Pandas`, `Matplotlib`.

**Фреймворк** - более "тяжёлое" решение, чем библиотека: это целая структура для построения приложений.
* Фреймворк задаёт **правила и архитектуру**, а программист «встраивает» свой код.
* Примеры: `Django` (веб-разработка), `TensorFlow` (машинное обучение).

**Решение "из коробки"** - готовый продукт или пакет, который почти не требует дополнительной настройки.
* Можно использовать сразу "как есть" для типовых задач.
* Примеры: CMS вроде `WordPress` (готовый сайт), аналитические платформы, AutoML-системы.

**Метаданные модуля:**
- `__builtins__` — ссылка на встроенные функции и объекты Python.  
- `__cached__` — путь к скомпилированному файлу `.pyc` в `__pycache__`.  
- `__doc__` — строка документации модуля (docstring) или `None`.  
- `__file__` — путь к файлу `.py`, из которого загружен модуль.  
- `__loader__` — объект-загрузчик модуля (используется при импорте).  
- `__name__` — имя модуля (`"__main__"` при прямом запуске).  
- `__package__` — имя пакета, которому принадлежит модуль.  
- `__spec__` — объект с информацией о загрузке модуля (ModuleSpec).

**Автоматическая перезагрузка модулей в Jupyter**

1. Подключите расширение:
   ```python
   %load_ext autoreload
````

2. Включите режим:

   * `%autoreload 0` — отключить перезагрузку.
   * `%autoreload 1` — перезагружать только модули из `%aimport`.
   * `%autoreload 2` — перезагружать все модули.

3. Управление модулями:

   * `%aimport module` — добавить модуль.
   * `%aimport -module` — исключить модуль.
   * `%aimport` — показать список.

`%aimport` — это магическая команда IPython, связанная с `%autoreload`.
Она управляет тем, **какие модули будут перезагружаться автоматически**.

*Основные варианты*

* `%aimport module` — добавить модуль в список отслеживаемых.
* `%aimport -module` — исключить модуль из списка.
* `%aimport` без аргументов — показать текущие списки:

  * *Modules to reload* — какие будут перезагружаться.
  * *Modules not to reload* — исключения.
  

* Если стоит `%autoreload 1`, то **автоматически перезагружаются только модули из списка `%aimport`**.
* Если стоит `%autoreload 2`, то перезагружаются все модули, **кроме исключённых через `%aimport -module`**.

`````python
%load_ext autoreload
%autoreload 2

import test_module
`````

**Работа с модулями из других директорий**

*Случай №1: модуль находится во внутренних директориях*
- Решение: просто прописываем названия папок последовательно через точки.

`````python
import folder.my_module
`````

*Случай №2: модуль находится в директориях выше*
- Решение: необходимо как-то "поднять" текущий путь "выше" по иерархии файловой системы.

**Идея:** Python ищет модули в списке директорий `sys.path`. Если модуль находится выше текущей папки, можно добавить путь на нужный уровень вручную:

```python
import sys

# Добавляем директорию на два уровня выше текущей
sys.path.append('../..')

# Теперь можно импортировать модуль
import my_module
```

`'../..'` означает: *подняться на два уровня вверх от текущей рабочей директории*.

Python превращает это в абсолютный путь и ищет там модуль.

`````python
! pwd
`````

`````python
import sys

sys.path.append('../..')

import my_module
`````

## 🟢 3. Настройка виртуального окружения.

**Зачем это нужно?**
Наш код протестирован только локально на нашем компьютере.
Что, если его запустит кто-то другой у себя?
- у него может не быть нужных библиотек
- у него может не быть нужных версий библиотек
- у него может быть другая, несовместимая ОС…

Это проблема **воспроизводимости**!


Решим пока проблему библиотек и их версий:)

Для этого можно создать **виртуальную среду, изолированную от основной системы** ⇒ каждый проект будет иметь свои независимые настройки, в т.ч. разные библиотеки и их версии.
```bash
$ pip install venv
$ python3 -m venv venv_name
$ source venv_name/bin/activate
(venv_name) $ pip install scikit-learn
deactivate

(venv_name)$ pip freeze
(venv_name) $ pip freeze > requirements.txt
pip install -r requirements.txt
```

**Сравните:**
```bash
$ which python3
```
и
```bash
(venv_name) $ which python3
```

При активации в.о. меняются специальные переменные среды ОС ⇒ **при вызове интерпретатора используется не глобальная версия интерпретатора Python, который имеет доступ ко всем установленным пакетам, а его копия, которая лежит в папке с в.о.**
Он имеет доступ только к “своей” папке, где хранятся библиотеки, установленные в это в.о.

**Привязка виртуального окружения к Jupyter:**
```bash
(venv_name)$ pip install ipykernel
(venv_name)$ python -m ipykernel install --user --name=venv_name --display-name "Python (venv_name)"
```
* `--name` — уникальное имя ядра в Jupyter.
* `--display-name` — то, как ядро будет отображаться в интерфейсе Jupyter Notebook.

Теперь в Jupyter можно выбрать это ядро, и ноутбук будет использовать пакеты из конкретного виртуального окружения.

## 🟢 4. Введение в git.
### Часть 0. Установка + настройка

**1. Установка Git**
```bash
# Ubuntu / Debian
$ sudo apt update
$ sudo apt install git

# MacOS (через Homebrew)
$ brew install git
# Проверка версии
$ git --version
```

Windows:
1. Скачайте установщик с официального сайта: https://git-scm.com/download/win
2. Запустите установщик и следуйте инструкциям (рекомендуется оставить все настройки по умолчанию).
3. После установки откройте **Git Bash** (или CMD/PowerShell) и проверьте версию: `$ git --version`

**2. Настройка Git (один раз)**

```bash
$ git config --global user.name "Ваше Имя"
$ git config --global user.email "your_email@example.com"
$ git config --global color.ui auto    # подсветка в терминале
```
----

### Часть 1. Удалённый репозиторий -> локальный
**1. Создание репозитория на GitHub**

1. Зайти на GitHub, выбрать New Repository.
2. Дать имя репозиторию, можно добавить README (опционально).
3. Скопировать URL для клонирования (HTTPS или SSH).

**2. Связывание локального репозитория с удалённым**

```bash
# Клонируем репозиторий
$ git clone https://github.com/username/repo_name.git

# Переходим в папку проекта
$ cd repo_name
```

**3. Создание первого файла и коммит**

```bash
$ echo "# Мой проект" > README.md
$ git add README.md       # добавляем файл в staging
$ git commit -m "Первый коммит"  # создаём коммит
```
**4. Генерация ключа:**

Для работы с удалёнными репозиториями GitHub требуется аутентификация.  
Наиболее удобный и безопасный способ — использовать **SSH-ключи**.  

SSH-ключи позволяют подключаться к репозиторию без постоянного ввода логина и пароля.  
Мы создаём пару ключей: **публичный**, который добавляется в GitHub, и **приватный**, который остаётся на вашем компьютере.  
После настройки SSH можно выполнять команды `git push` и `git pull` автоматически, без ручного ввода данных.

- Сгенерируем ключ:
```bash
$ ssh-keygen -t ed25519 -C "your_email@example.com"
# или для старых систем: -t rsa -b 4096
```

- Добавляем публичный ключ в GitHub:
   Settings -> SSH and GPG keys -> New SSH key -> вставить содержимое `~/.ssh/id_ed25519.pub`.

- Клонируем через SSH:

```bash
$ git clone git@github.com:username/repo_name.git
```

- После этого `git push`/`git pull` не требует ввода логина/пароля.

**4. Отправка на GitHub**

```bash
$ git push origin main  # отправка изменений в ветку main на GitHub
```

### Часть 2. Локальный репозиторий -> удалённый репозиторий

**1. Создание нового локального репозитория**

```bash
$ mkdir my_local_repo
$ cd my_local_repo
$ git init  # инициализация git
```

**2. Добавление файлов и первый коммит**

```bash
$ echo "print('Hello Git')" > hello.py
$ git add hello.py
$ git commit -m "Первый коммит в локальный репо"
```

**3. Проверка статуса**

```bash
$ git status   # показывает, какие файлы изменены или в staging
$ git log      # история коммитов
```

**4. Дальнейшая работа**

* Добавляем/редактируем файлы → `git add <file>` -> `git commit -m "сообщение"`
* Можно создавать ветки:

```bash
$ git branch new_feature
$ git checkout new_feature
```

**5. Связывание с удалённым репозиторием***

* Создайте пустой репозиторий на GitHub (не добавляя README).  
* Добавьте удалённый репозиторий к локальному:

```bash
$ git remote add origin git@github.com:username/repo_name.git  # через SSH
# или через HTTPS:
# $ git remote add origin https://github.com/username/repo_name.git
````

* Проверьте, что удалённый репозиторий добавлен:

```bash
$ git remote -v
origin  git@github.com:username/repo_name.git (fetch)
origin  git@github.com:username/repo_name.git (push)
```

* Можно отправлять коммиты на GitHub!

```bash
$ git push origin main  # отправка ветки main в удалённый репозиторий
```
