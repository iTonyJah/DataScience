### Реализация алгоритма

Так как данные носят категориальный характер, то для реализции будем использовать библиотеку skikit-learn CategoricalNB, которая ожидает, что категории представлены в целых числах, начиная с 0. Для этого, предварительно, преобразуем текстовые  в числовые.

```python
# Создание рабочей копии с подготовленной таблицы
data = df.copy()

# Преобразование в числа
data['Age_Group'] = data['Age_Group'].map({'Young': 0, 'Old': 1})
data['Test'] = data['Test'].map({'Negative': 0, 'Positive': 1})
data['Status'] = data['Status'].map({'Not_infected': 0, 'Infected': 1})
```

```python
# Разделение данных
X_train, X_test, y_train, y_test = train_test_split(
    data[['Age_Group', 'Test']], 
    data['Status'], 
    test_size=0.2,
    random_state=42
)
```

```python
# Создание и обучение модели
model = CategoricalNB()
model.fit(X_train, y_train)

# Предсказание на тестовых данных
y_pred = model.predict(X_test)

# Оценка точности
accuracy = model.score(X_test, y_test)
print(f"Точность модели: {accuracy:.2%}")

# Пример предсказания
new_person = pd.DataFrame({'Age_Group': [0], 'Test': [1]})  # Young, Positive
prediction = model.predict(new_person)
print(f"Пример предсказания: Молодой с положительным тестом: {'Заболел' if prediction[0]==1 else 'Здоров'}")
```

```python
# точные значения предсказания
model.predict_proba(new_person)
```

```python
# Вывод основных метрик
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2%}")
print(f"Precision: {precision_score(y_test, y_pred):.2%}")
print(f"Recall: {recall_score(y_test, y_pred):.2%}")
```

$ Accuracy = \frac{P}{N} = \frac {TP + TN} {TP + TN + FP + FN}$ ,
$ Precision= \frac {TP}{TP+FP}$ ,
$ Recall= \frac {TP}{TP+FN} $

* TP — истино-положительное решение;
* TN — истино-отрицательное решение;
* FP — ложно-положительное решение;
* FN — ложно-отрицательное решение.

```python
# Вычисляем матрицу ошибок
cm = confusion_matrix(y_test, y_pred)

# Создаем DataFrame с подписями
cm_df = pd.DataFrame(
    cm,
    index=['Фактически: Здоров', 'Фактически: Заражен'],
    columns=['Предсказано: Здоров', 'Предсказано: Заражен']
)

print("Матрица ошибок:")
print(cm_df)
```

### Реализация алгоритма с нуля

```python
class NaiveBayesFromScratch:
    """
    Наивный байесовский классификатор для категориальных признаков.
    Предполагает условную независимость признаков при данном классе.
    """    
    def fit(self, X, y):
        """
        Обучение модели на обучающих данных.
        Вычисляет априорные вероятности классов и условные вероятности признаков.
        """
        # Определяем уникальные классы
        self.classes = y.unique()
        # Словари для хранения априорных вероятностей P(Status)
        self.prior_probs = {}
        # Словари для хранения условных вероятностей P(Feature|Status)
        # Структура: {Feature_Name: {Status_Value: {Feature_Value: Probability}}}
        self.likelihoods = {}

        for cls in self.classes:
            # Расчет априорной вероятности P(класс)
            # Формула: P(y) = (количество объектов класса y) / (общее количество)
            # Считаем общее количество каждой категории статуса и делим на общее количество строк.
            self.prior_probs[cls] = len(y[y == cls]) / len(y)
            self.likelihoods[cls] = {}
            # Перебираем каждый признак (Age_Group, Test)
            for feature in X.columns:
                self.likelihoods[cls][feature] = {}
                # Расчет вероятности признака при условии класса P(признак|класс)
                for value in X[feature].unique():
                    count = len(X[(y == cls) & (X[feature] == value)])
                    total_count = len(X[y == cls])
                    # Используем сглаживание Лапласа (добавление 1) для предотвращения нулевых вероятностей
                    # Если данных много, можно обойтись без сглаживания в простейшем случае
                    self.likelihoods[cls][feature][value] = (count + 1) / (total_count + len(X[feature].unique()))

    def predict(self, X):
        """
        Предсказание вероятностей принадлежности к классам.
        Возвращает список предсказанных значений для X.
        """        
        predictions = []
        for index, row in X.iterrows():
            posterior_probs = {}
            for cls in self.classes:
                # Начинаем с априорной вероятности P(класс)
                posterior_probs[cls] = self.prior_probs[cls]
                # Умножаем на вероятности признаков P(признак|класс)
                for feature, value in row.items():
                    if value in self.likelihoods[cls][feature]:
                        posterior_probs[cls] *= self.likelihoods[cls][feature][value]
                    else:
                        # Если значение признака не встречалось в обучении для этого класса,
                        # используем минимальную вероятность или ноль (в примере со сглаживанием это уже учтено)
                        posterior_probs[cls] *= 1 / len(self.likelihoods[cls][feature].keys())

            # Выбираем класс с наибольшей апостериорной вероятностью
            best_class = max(posterior_probs, key=posterior_probs.get)
            predictions.append(best_class)
        return predictions

# Обучение модели
nb_classifier = NaiveBayesFromScratch()
nb_classifier.fit(X_train, y_train)

# Предсказание на тестовой выборке
predictions = nb_classifier.predict(X_test)
```

```python
# Оценка точности (Accuracy)
accuracy = accuracy_score(y_test, predictions)
print(f"Точность модели (accuracy): {accuracy:.4f}\n")

# Подробный отчет по классификации (Precision, Recall, F1-score)
print("Отчет по классификации (расчёт библиблиотеки):")
print(classification_report(y_test, predictions))
```

Значения метрики *accuracy* (69.64%) совпадают.

```python
# Новый набор данных для предсказания
new_data = pd.DataFrame({'Age_Group': [0], 'Test': [1]})

# Получение предсказания
prediction_for_new_data = nb_classifier.predict(new_data)

print(f"\nДанные для нового пациента:\n{new_data}")
print(f"\nПредсказанный статус (модель Наивного Байеса): {prediction_for_new_data[0]}")
print(f"Пример предсказания: Молодой с положительным тестом: {'Заболел' if prediction_for_new_data[0]==1 else 'Здоров'}")
```

Предсказания совпадают.

```python
# Самостоятельно рассчитаем метрики
#
# Преобразуем Series и list в numpy массивы для удобства сравнения
y_true = np.array(y_test)
y_pred_manual = np.array(predictions)

def calculate_metrics_manual(y_true, y_pred, positive_label=1):
    # Определяем компоненты матрицы ошибок
    TP = np.sum((y_true == positive_label) & (y_pred == positive_label))
    TN = np.sum((y_true != positive_label) & (y_pred != positive_label))
    FP = np.sum((y_true != positive_label) & (y_pred == positive_label))
    FN = np.sum((y_true == positive_label) & (y_pred != positive_label))

    # Избегаем деления на ноль, если знаменатель равен 0
    accuracy = (TP + TN) / len(y_true) if len(y_true) > 0 else 0
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0
    f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    return {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1_score,
        "TP": TP, "TN": TN, "FP": FP, "FN": FN
    }

# Рассчитываем метрики, считая класс 1 положительным
metrics = calculate_metrics_manual(y_true, y_pred_manual, positive_label=1)

print("\n--- Результаты ручного расчета метрик ---")
print(f"TP (Истинно-положительные): {metrics['TP']}")
print(f"TN (Истинно-отрицательные): {metrics['TN']}")
print(f"FP (Ложно-положительные): {metrics['FP']}")
print(f"FN (Ложно-отрицательные): {metrics['FN']}")
print("-" * 35)
print(f"Accuracy (Точность): {metrics['Accuracy']:.4f}")
print(f"Precision (Точность предсказания): {metrics['Precision']:.4f}")
print(f"Recall (Полнота обнаружения): {metrics['Recall']:.4f}")
print(f"F1-Score: {metrics['F1-Score']:.4f}")
```

Метрики модели совпадают с библиотечным вариантом.

## Задание 5

Дана функция:

$f(x)=x^{4}+3 x^{3}-12 x^{2}+7 x-2$

Оптимизируйте её с помощью метода Ньютона, используя язык программирования Python.

За начальную точку возьмите х = 10, необходимую точность 0.0001.



### Решение:

```python
def f(x):
    """Возвращает значение исходной функции"""
    return x**4 + 3*x**3 - 12*x**2 + 7*x - 2

def df(x):
    """Возвращает значение первой производной"""
    return 4*x**3 + 9*x**2 - 24*x + 7

def ddf(x):
    """Возвращает значение второй производной"""
    return 12*x**2 + 18*x - 24

def newton_optimization(initial_x, tolerance):
    """
    Метод Ньютона для оптимизации (нахождения экстремума функции)
   
    Возвращает:
    x_new - найденная точка экстремума
    iteration - количество выполненных итераций
    """
    x = initial_x
    iteration = 0
    while True:
        df_val = df(x)
        ddf_val = ddf(x)

        print(f"Итерация {iteration}: x = {x:.3f} / f'(x) = {df_val:.3f} /  f''(x) = {ddf_val:.3f}")
        
        if ddf_val == 0:
            print("Вторая производная равна нулю. Метод не может продолжить.")
            break
        
        x_new = x - df_val / ddf_val
        if abs(x_new - x) < tolerance:
            break
        x = x_new
        iteration += 1
        
    return x_new, iteration

# Начальная точка и точность
initial_x = 10
tolerance = 0.0001

# Запуск оптимизации
optimal_x, iterations = newton_optimization(initial_x, tolerance)

print(f"Оптимальное значение x: {optimal_x}")
print(f"Значение функции в этой точке: f(x) = {f(optimal_x):.8f}")
print(f"Количество итераций: {iterations}")

# Дополнительная информация о характере экстремума
fdp_result = ddf(optimal_x)
if fdp_result > 0:
    print(f"Так как f''(x) = {fdp_result:.3f} > 0, это точка минимума")
elif fdp_result < 0:
    print(f"Так как f''(x) = {fdp_result:.3f} < 0, это точка максимума")
else:
    print("Требуется дополнительный анализ (вторая производная равна нулю)")
```

### Ответ:

Минимум функции (-3,73) находится в точке 1,31.
