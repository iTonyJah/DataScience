Вычислите результат самостоятельно (вручную) и с помощью Python. Сравните результаты.

## Самостоятельно вычисление (вручную):


![условные экстремумы](1_dz_2.jpg)


```python
# сохраним полученные занчения для сравнения
sols_man = [{a: -3, b: -1, w: -2}]
sols_man.append({a: 3, b: 1, w: 2})
```

```python
sols_man
```

## Вычисление с помощью Python:


```python
from sympy import *
a, b, w = symbols('a b w' )

g = 4*a + 8*b
print('Целевая функция для аргументов a и b :\n f = ', g)

q = b**2 - 2*a*b + 5
print('Функция ограничений: ', q,'= 0')

f = 4*a + 8*b + w * (b**2 - 2*a*b + 5)
print('Функция Лагранжа :\n ', f)

fa = f.diff(a)
print('df/da =', fa,'= 0')

fb = f.diff(b)
print('df/db =', fb,'= 0')

fw = f.diff(w)
print('df/dw =', fw,'= 0')

sols = solve([fa, fb, fw], (a, b, w), dict=True)
print('Стационарная точка M(x,y)_1:\n',float(sols[0][a]),',',float(sols[0][b]))
print('Стационарная точка M(x,y)_2:\n',float(sols[1][a]),',',float(sols[1][b]))
```

## Сравнение результатов:

```python
#сравним значения рассчитанные с помощью Python (слева) и расчитанные вручную
print(sols[0][a] == sols_man[0][a])
print(sols[0][b] == sols_man[0][b])
print(sols[0][w] == sols_man[0][w])
print(sols[1][a] == sols_man[1][a])
print(sols[1][b] == sols_man[1][b])
print(sols[1][w] == sols_man[1][w])
```

**Ответ:** Условные экстремумы функции находятся:
* в точке $(-3, -1)$ - условный минимум, со значением функции в этой точке равным $-20$
* и в точке $(3, 1)$ - условный максимум, значение функции в этой точке $20$.

Значения рассчитаные вручную и с помощью Python - одинаковые.

Вам предложены данные с информацией о успеваемости студентов:

Независимые переменные:
* Hours Studied: Общее количество часов, потраченных на учебу каждым студентом.
* Previous Scores: Баллы, полученные студентами на предыдущих экзаменах.
* Sleep Hours: Среднее количество часов сна студента в сутки.
* Sample Question Papers Practiced: Количество пробных экзаменационных работ, с которыми студент занимался.
  
Целевая переменная:
* Performance Index: Показатель общей успеваемости каждого студента. Индекс успеваемости отражает академическую успеваемость студента и округляется до ближайшего целого числа. Индекс варьируется от 10 до 100, при этом более высокие значения свидетельствуют о более высокой успеваемости.

Решите задачу линейной регрессии, реализовав градиентный спуск самостоятельно, не используя готовое решение из какой-либо библиотеки.

```python
import numpy as np
import pandas as pd

df = pd.read_csv('Student_Performance.txt')

df
```

```python
X = df[['Hours Studied', 'Previous Scores', 'Sleep Hours', 'Sample Question Papers Practiced']].values

y = df['Performance Index']
n = len(y)
X = np.append(np.ones((n, 1)), X, axis=1)

y = df['Performance Index'].values.reshape(n, 1)

par = np.zeros((5, 1))

def performance_index(X, y, par):
    y_pred = np.dot(X, par)
    error = (y_pred - y)**2
    perfect = (1/(2*n)) * np.sum(error)

    return perfect


def grad_d(X, y, par, alpha, iterations):
    perf_ind = []
    for i in range(iterations):
        y_pred = np.dot(X, par)
        der = np.dot (X.transpose(), (y_pred - y))/ n
        par -= alpha * der
        perf_ind.append(performance_index(X, y, par))

        if i % 50_000 == 0:
            print(f"iter={i}", par[0], par[1], par[2], par[3], par[4])

    print(f"iter={i}", par[0], par[1], par[2], par[3], par[4])    
    return par, perf_ind

# 0.000384 - максимально возможный шаг, на бОльших значениях начинает вылетать
step = 0.000384

# меньше итераций, более приемлимое время, но результат не такой точный, только до первых трёх знаков
iter = 700_000

# долго считает, но выдаёт результат равный библиотечному
# iter = 2_300_000

par, perf_opt = grad_d(X, y, par, step, iter)
```
