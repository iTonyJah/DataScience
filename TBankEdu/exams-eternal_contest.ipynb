У Артемия есть бесконечное число монет, каждая из которых одного из трех номиналов. Его интересует, какие суммы от 1 до N рублей он может набрать в свой кошелек, если там заранее лежала монета величиной в 1 рубль.

Формат входных данных
Первая строка содержит число N — ограничение на сумарную стоимость монет в кошельке (1 <= N <= 10^18).
Вторая строка содержит три числа A, B, C задающие достоинства типов монет (1 <= A, B, C <= 100000).

Формат выходных данных
Выведите единственное число — количество сумм, которые можно набрать в кошельке. 

```python
import sys
import math

def solve():
    N = int(sys.stdin.readline())
    A, B, C = map(int, sys.stdin.readline().split())
    
    # Проверяем, что gcd(A, B, C) делит 1 (но у нас уже есть 1, так что это всегда верно)
    g = math.gcd(math.gcd(A, B), C)
    if g != 1:
        print(1 if 1 <= N else 0)
        return
    
    max_coin = max(A, B, C)
    # Динамика: can[i] = можно ли набрать сумму i
    can = [False] * (2 * max_coin + 2)
    can[0] = True
    for i in range(1, 2 * max_coin + 1):
        if (i >= 1 and can[i - 1]) or \
           (i >= A and can[i - A]) or \
           (i >= B and can[i - B]) or \
           (i >= C and can[i - C]):
            can[i] = True
    
    # Находим все недостижимые числа до 2 * max_coin
    missing = []
    for i in range(1, 2 * max_coin + 1):
        if not can[i]:
            missing.append(i)
    
    # Если N <= 2 * max_coin, просто считаем сколько можно
    if N <= 2 * max_coin:
        res = N - len([x for x in missing if x <= N])
        print(res)
        return
    
    # Иначе все числа >= M можно набрать, где M - минимальное после которого все достижимы
    M = 2 * max_coin
    res = N - len(missing)
    print(res)

solve()
```
