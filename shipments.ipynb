```python
sh = sh_main_3m_xlsx.copy()
```

```python
sh.shape
```

```python
if sh.iloc[-1,0] == 'Итого':
    print('drop Итого')
    sh.drop(sh.tail(1).index, inplace=True)
```

```python
# sh.to_pickle('sh12m.pkl')
```

```python
sh['Дата доставки'] = pd.to_datetime(sh['Дата доставки'], dayfirst=True)
sh['Дата сбора'] = pd.to_datetime(sh['Дата сбора'], dayfirst=True)
sh['period'] = sh['Дата сбора'].dt.strftime('%Y-%m')
```
